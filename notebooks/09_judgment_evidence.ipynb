```python
# ==========================================================================
# SELF-CONTAINED NOTEBOOK (single cell). The project's tested src/ modules
# are embedded (base64) and loaded in-memory, so this notebook runs on its
# own. Put the AI Hub zips in data/raw/ and your key in .env as usual.
# ==========================================================================
import sys, types, base64
from pathlib import Path
_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
_SRC = {
    'config': 'IiIiCmNvbmZpZy5weSDigJQg7ZSE66Gc7KCd7Yq4IOyghOyXrSDshKTsoJUgKOqyveuhnCDCtyDsi5zrk5wgwrcg7IOB7IiYKQoK7J6s7ZiE7ISxIOybkOy5mToKICAtIOuqqOuToCDqsr3roZzripQg7J20IO2MjOydvOydmCDsnITsuZjrpbwg6riw7KSA7Jy866GcIOqzhOyCsO2VnOuLpCjtlZjrk5zsvZTrlKkg7KCI64yA6rK966GcIOq4iOyngCkuCiAgLSDrqqjrk6Ag66y07J6R7JyEIOyLnOuTnOuKlCBTRUVEIO2VmOuCmOuhnCDthrXsnbztlZzri6QuCiAgLSDsubTthYzqs6Drpqwv652867KoIOyDgeyImOuKlCDsl6zquLDshJzrp4wg7KCV7J2Y7ZWY6rOgIOuFuO2KuOu2geydgCBpbXBvcnQg7ZW07IScIOyTtOuLpC4KIiIiCmZyb20gX19mdXR1cmVfXyBpbXBvcnQgYW5ub3RhdGlvbnMKaW1wb3J0IG9zCmltcG9ydCByYW5kb20KZnJvbSBwYXRobGliIGltcG9ydCBQYXRoCgojIC0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tCiMg6rK966GcICjsnbQg7YyM7J28OiA8UFJPSkVDVF9ST09UPi9zcmMvY29uZmlnLnB5IOq4sOykgCkKIyAtLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLQpQUk9KRUNUX1JPT1QgPSBQYXRoKF9fZmlsZV9fKS5yZXNvbHZlKCkucGFyZW50c1sxXQoKREFUQV9ESVIgICAgICA9IFBST0pFQ1RfUk9PVCAvICJkYXRhIgpSQVdfRElSICAgICAgID0gREFUQV9ESVIgLyAicmF3IiAgICAgICAgICAjIEFJIEh1YiDsm5Drs7ggemlwICjsnb3quLAg7KCE7JqpLCDsiJjsoJUg6riI7KeAKQpJTlRFUklNX0RJUiAgID0gREFUQV9ESVIgLyAiaW50ZXJpbSIgICAgICAjIO2MjOyLscK37KCV7KCcIOykkeqwhCDsgrDstpzrrLwKUFJPQ0VTU0VEX0RJUiA9IERBVEFfRElSIC8gInByb2Nlc3NlZCIgICAgIyDsmKjthqjroZzsp4Ag64W465OcL+yXo+yngCwg7KSA7IiY7YyQ7KCVIOqzqOuTnAoKUkVTVUxUU19ESVIgICA9IFBST0pFQ1RfUk9PVCAvICJyZXN1bHRzIgpGSUdVUkVTX0RJUiAgID0gUkVTVUxUU19ESVIgLyAiZmlndXJlcyIKVEFCTEVTX0RJUiAgICA9IFJFU1VMVFNfRElSIC8gInRhYmxlcyIKCmZvciBfZCBpbiAoSU5URVJJTV9ESVIsIFBST0NFU1NFRF9ESVIsIEZJR1VSRVNfRElSLCBUQUJMRVNfRElSKToKICAgIF9kLm1rZGlyKHBhcmVudHM9VHJ1ZSwgZXhpc3Rfb2s9VHJ1ZSkKCiMgLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0KIyDsnqztmITshLE6IOyLnOuTnAojIC0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tClNFRUQgPSA0MgoKZGVmIHNldF9zZWVkKHNlZWQ6IGludCA9IFNFRUQpIC0+IE5vbmU6CiAgICAiIiLrqqjrk6Ag65287J2067iM65+s66asIOyLnOuTnOulvCDtlZwg67KI7JeQIOqzoOyglS4g64W47Yq467aBIOyyqyDshYDsl5DshJwg7Zi47LacLiIiIgogICAgcmFuZG9tLnNlZWQoc2VlZCkKICAgIG9zLmVudmlyb25bIlBZVEhPTkhBU0hTRUVEIl0gPSBzdHIoc2VlZCkKICAgIHRyeToKICAgICAgICBpbXBvcnQgbnVtcHkgYXMgbnAKICAgICAgICBucC5yYW5kb20uc2VlZChzZWVkKQogICAgZXhjZXB0IEltcG9ydEVycm9yOgogICAgICAgIHBhc3MKCiMgLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0KIyDsm5Dsspwg642w7J207YSwIOyDgeyImAojIC0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tCiMgQUkgSHViIOybkOyynCB6aXAg7YyM7J2866qFIChkYXRhL3Jhdy8g7JWE656Y7JeQIOq3uOuMgOuhnCDsnITsuZgpClNPVVJDRV9aSVBTID0gWwogICAgIlRTXzFfX+uyoO2KuOuCqF8wMV9f6rec7KCcLnppcCIsCiAgICAiVFNfMV9f67Kg7Yq464KoXzAyX1/snoXrspVf64+Z7ZalLnppcCIsCiAgICAiVFNfMV9f67Kg7Yq464KoXzAzX1/tjJDroYAuemlwIiwKICAgICJUU18xX1/rsqDtirjrgqhfMDRfX+q4sOyCrC56aXAiLAogICAgIlRTXzFfX+uyoO2KuOuCqF8wNV9f7KCV67aAX+uwj1/quLDsl4VfRVNHX+yekOujjC56aXAiLAogICAgIlRTXzFfX+uyoO2KuOuCqF8wNl9f6rWt7KCcX0VTR1/snpDro4wuemlwIiwKICAgICJUU18yX1/rp5DroIjsnbTsi5zslYRfMDFfX+q3nOygnC56aXAiLAogICAgIlRTXzJfX+unkOugiOydtOyLnOyVhF8wMl9f7J6F67KVX+uPme2WpS56aXAiLAogICAgIlRTXzJfX+unkOugiOydtOyLnOyVhF8wM19f7YyQ66GALnppcCIsCiAgICAiVFNfMl9f66eQ66CI7J207Iuc7JWEXzA0X1/quLDsgqwuemlwIiwKICAgICJUU18yX1/rp5DroIjsnbTsi5zslYRfMDVfX+ygleu2gF/rsI9f6riw7JeFX0VTR1/snpDro4wuemlwIiwKICAgICJUU18yX1/rp5DroIjsnbTsi5zslYRfMDZfX+q1reygnF9FU0df7J6Q66OMLnppcCIsCl0KTEFCRUxfWklQID0gIjAyX+udvOuyqOungeuNsOydtO2EsC56aXAiCgpDQVRFR09SSUVTID0gWyLqt5zsoJwiLCAi7J6F67KVIOuPme2WpSIsICLtjJDroYAiLCAi6riw7IKsIiwgIuygleu2gCDrsI8g6riw7JeFIEVTRyDsnpDro4wiLCAi6rWt7KCcIEVTRyDsnpDro4wiXQpDT1VOVFJJRVMgID0gWyLrsqDtirjrgqgiLCAi66eQ66CI7J207Iuc7JWEIl0KRVNHX1RZUEVTICA9IFsiRSIsICJTIiwgIkciXQoKIyDsi6TsuKEg6riw7KSA6rCSIChwYXJzaW5nIOqygOymneyaqSDigJQg66Gc7Lus7JeQ7IScIOydtCDqsJLqs7wg64us65287KeA66m0IOuNsOydtO2EsCDriITrnb0g6rK96rOgKQpFWFBFQ1RFRF9TT1VSQ0VfVE9UQUwgPSA0MDA3OCAgICAgICAgICAjIOyXheuhnOuTnCDsoJzqs7XrtoQgKOqzteyLnSDsoITssrQgNTAsMDk4KQpFWFBFQ1RFRF9DQVRFR09SWV9DT1VOVFMgPSB7CiAgICAi6rec7KCcIjogOTYxNCwgIuq1reygnCBFU0cg7J6Q66OMIjogNzk4NiwgIuygleu2gCDrsI8g6riw7JeFIEVTRyDsnpDro4wiOiA3OTUxLAogICAgIuyeheuylSDrj5ntlqUiOiA3MjM0LCAi6riw7IKsIjogNDc2NCwgIu2MkOuhgCI6IDI1MjksCn0KRVhQRUNURURfQ09VTlRSWV9DT1VOVFMgPSB7IuuyoO2KuOuCqCI6IDIwMDc5LCAi66eQ66CI7J207Iuc7JWEIjogMTk5OTl9CkVYUEVDVEVEX0VTR19DT1VOVFMgPSB7IkUiOiAyNjM0MiwgIlMiOiA5MTA0LCAiRyI6IDQ2MzJ9CgojIC0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tCiMg7KSA7IiY7YyQ7KCVIOudvOuyqCDsg4HsiJggKGNvbXBsaWFuY2UgbGFiZWxpbmcpCiMgLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0KTEFCRUxfTk9OX0NPTVBMSUFOVCA9ICJOT05fQ09NUExJQU5UIgpMQUJFTF9DT01QTElBTlQgICAgID0gIkNPTVBMSUFOVCIKTEFCRUxfT1RIRVIgICAgICAgICA9ICJPVEhFUiIgICAgICAjIFBBUlRJQUwvTUlYRUQv67aI66qFIOKGkiDsnbjsoIEg6rKA7IiYCkxBQkVMX1VOTEFCRUxFRCAgICAgPSAiVU5MQUJFTEVEIiAgIyByZXN1bHQg6rO1656ACgojIOygnOyerCDsi6DtmLg6IOuqhe2Zle2VnCDsspjrsowg6rKw6rO8IOyWtO2cmC4KIyDso7zsnZgpIO2ZlO2PkCDri6jsnIQgIuuPmSIoVk5EKeydgCAn7YyQ6rKw67mE7JqpIOu2gOuLtCcg65OxIOykkeumvSDrrLjqtazsl5Drj4Qg65Ox7J6l7ZWY7JesCiMgICAgICAg6rO864ukIO2PrO2ajeydhCDsnKDrsJztlZjrr4DroZwgU0FOQ1RJT04g7Yag7YGw7JeQ7IScIOygnOyZuO2VnOuLpCjCp2xhYmVsaW5nIOyjvOyEnSDssLjsobApLgpTQU5DVElPTl9UT0tFTlMgPSBbCiAgICAi7KeV7JetIiwgIuuyjOq4iCIsICLtg5ztmJUiLCAi7IaQ7ZW067Cw7IOBIiwgIuuwsOyDgeyxheyehCIsICLtlbTqs6AiLAogICAgIuqzvO2DnOujjCIsICLrqrDsiJgiLCAi7JiB7JeF7KCV7KeAIiwgIuuTseuhney3qOyGjCIsICLst6jshowiLCAi7ZmY7IiYIiwKXQpBQ1FVSVRfVE9LRU5TID0gWwogICAgIuustOyjhCIsICLquLDqsIEiLCAi6rCB7ZWYIiwgIuuptOygnCIsICLssYXsnoQg7JeG7J2MIiwgIu2VtOuLuSDsl4bsnYwiLCAi7JeG7J2MIiwKXQoKCiMgPT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT0KIyBSRVZJU0lPTiBBREQtT05TIChSMSByZXZpc2lvbiwgMjAyNi0xMCkg4oCUIG5ldyBleHBlcmltZW50IGNvbnN0YW50cwojIC0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tCiMgS2VwdCBhcHBlbmQtb25seSBzbyBub3RlYm9va3MgMDEtMDUgYmVoYXZlIGlkZW50aWNhbGx5IHRvIHRoZSBzdWJtaXR0ZWQgcnVuLgojID09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09CgojIC0tLS0gZGV0ZXJtaW5pc3RpYyB0cmFpbi92YWwvdGVzdCBzcGxpdCBmb3IgaGVsZC1vdXQgaHlwZXItcGFyYW1ldGVyIHR1bmluZwpTUExJVF9GUkFDUyA9IHsidHJhaW4iOiAwLjUsICJ2YWwiOiAwLjI1LCAidGVzdCI6IDAuMjV9ICAjIGJ5IHF1ZXJ5CgojIC0tLS0gcmV0cmlldmFsIGNvcnB1cyAvIGV2YWwKRE9DX0NIQVJfTElNSVQgPSAzMDAwICAgICAgICAgICMgc2FtZSBoZWFkIHdpbmRvdyBhcyBuYjA0IChrZWVwIGNvbXBhcmFibGUpClJFVFJJRVZBTF9LX0xJU1QgPSAoMSwgNSwgMTApCkJPT1RTVFJBUF9OID0gMTAwMCAgICAgICAgICAgICAjIHJlc2FtcGxlcyBmb3IgQ0lzCkJPT1RTVFJBUF9DSSA9IDAuOTUKCiMgLS0tLSBkZW5zZSBtdWx0aWxpbmd1YWwgYmFzZWxpbmVzIChkb3dubG9hZGVkIG9uIHRoZSB1c2VyJ3MgbWFjaGluZSkKREVOU0VfTU9ERUxTID0gewogICAgIm1FNSI6ICAgICJpbnRmbG9hdC9tdWx0aWxpbmd1YWwtZTUtYmFzZSIsCiAgICAiQkdFLU0zIjogIkJBQUkvYmdlLW0zIiwKfQoKIyAtLS0tIFF1YWxpZmllZCBPbnRvbG9neSBBbmNob3JpbmcgKFFPQSk6IHRoZSBuLWFyeSBleHRlbnNpb24KIyBUcmlwbGUgICA6IChzdWJqZWN0LCBwcmVkaWNhdGUsIG9iamVjdCkgICAgICAgICAgICAgICAgICAgICAgLT4gQjMgbWV0YWRhdGEgYm9vc3QgKGp1cmlzZGljdGlvbiwgdG9waWMpCiMgUXVhZCAgICAgOiB0cmlwbGUgKyB0ZW1wb3JhbCB2YWxpZGl0eSBxdWFsaWZpZXIgICAgICAgICAgICAgIC0+ICsgZWZmZWN0aXZlL3Zpb2xhdGlvbi1kYXRlIGNvbnNpc3RlbmN5CiMgUGVudHVwbGUgOiBxdWFkICAgKyBzZXZlcml0eS9zdWJ0b3BpYyBxdWFsaWZpZXIgICAgICAgICAgICAgIC0+ICsgc2FuY3Rpb24tc2V2ZXJpdHkgLyBFU0ctc3VidG9waWMKUU9BX0RJTUVOU0lPTlMgPSBbImp1cmlzZGljdGlvbiIsICJ0b3BpYyIsICJ0ZW1wb3JhbCIsICJzZXZlcml0eSIsICJkb2N0eXBlIl0KUU9BX0xFVkVMUyA9IHsKICAgICJ0cmlwbGUiOiAgIFsianVyaXNkaWN0aW9uIiwgInRvcGljIl0sCiAgICAicXVhZCI6ICAgICBbImp1cmlzZGljdGlvbiIsICJ0b3BpYyIsICJ0ZW1wb3JhbCJdLAogICAgInBlbnR1cGxlIjogWyJqdXJpc2RpY3Rpb24iLCAidG9waWMiLCAidGVtcG9yYWwiLCAic2V2ZXJpdHkiLCAiZG9jdHlwZSJdLAp9CiMgdGVtcG9yYWwgZGVjYXkgY29uc3RhbnQgKHllYXJzKSBmb3IgZXhwKC18ZHR8L3RhdSkgcHJveGltaXR5ClFPQV9URU1QT1JBTF9UQVUgPSAzLjAKIyBjb29yZGluYXRlLWFzY2VudCBncmlkIGZvciBwZXItZGltZW5zaW9uIHdlaWdodHMgYWxwaGFfayAodHVuZWQgb24gdmFsIHNwbGl0KQpRT0FfQUxQSEFfR1JJRCA9IFswLjAsIDAuMSwgMC4yNSwgMC41LCAxLjAsIDIuMF0KCiMgLS0tLSBzYW5jdGlvbiBzZXZlcml0eSBidWNrZXRzIChvcmRpbmFsKSBmb3IgdGhlIHBlbnR1cGxlIHF1YWxpZmllcgpTRVZFUklUWV9PUkRFUiA9IFsiTk9ORSIsICJNSU5PUiIsICJNQUpPUiIsICJDUklNSU5BTCJdClNFVkVSSVRZX01BUCA9IHsKICAgICMgY3JpbWluYWwgLyBjdXN0b2RpYWwKICAgICLsp5Xsl60iOiAiQ1JJTUlOQUwiLCAi7YOc7ZiVIjogIkNSSU1JTkFMIiwKICAgICMgbWFqb3IgYWRtaW5pc3RyYXRpdmUgLyBmaW5hbmNpYWwKICAgICLrsozquIgiOiAiTUFKT1IiLCAi6rO87YOc66OMIjogIk1BSk9SIiwgIuyGkO2VtOuwsOyDgSI6ICJNQUpPUiIsICLrsLDsg4HssYXsnoQiOiAiTUFKT1IiLAogICAgIuyYgeyXheygleyngCI6ICJNQUpPUiIsICLrk7HroZ3st6jshowiOiAiTUFKT1IiLCAi66qw7IiYIjogIk1BSk9SIiwgIu2ZmOyImCI6ICJNQUpPUiIsICLtlbTqs6AiOiAiTUFKT1IiLAogICAgIyBtaW5vciAvIGNvcnJlY3RpdmUKICAgICLst6jshowiOiAiTUlOT1IiLCAi7Iuc7KCVIjogIk1JTk9SIiwgIuqyveqzoCI6ICJNSU5PUiIsCn0KCiMgLS0tLSBzZWxlY3RpdmUgcHJlZGljdGlvbgpTRUxFQ1RJVkVfUklTS19MRVZFTFMgPSBbMC4wNSwgMC4xMCwgMC4yMF0gICAjIHRhcmdldCByaXNrIHBvaW50cyBmb3IgY292ZXJhZ2VAcmlzawpUT1NUX01BUkdJTiA9IDAuMDUgICAgICAgICAgICAgICAgICAgICAgICAgICAgIyBlcXVpdmFsZW5jZSBtYXJnaW4gZm9yICJuZWdsaWdpYmxlIGdhcCIKCgojID09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09CiMgUkVWSVNJT04gQURELU9OUyAoUjEpOiBjYW5vbmljYWwgRW5nbGlzaCBjYXRlZ29yeS9jb3VudHJ5IG5hbWVzICsgS08tPkVOIG1hcHMKIyBOb3RlYm9vayAwMSBub3JtYWxpc2VzIHRoZSBwYXJzZWQgZGF0YSB0byB0aGVzZSBFbmdsaXNoIG5hbWVzIGJlZm9yZSBzYXZpbmcsCiMgc28gZXZlcnkgZG93bnN0cmVhbSBub3RlYm9vayBmaWx0ZXJzIG9uIEVuZ2xpc2ggYW5kIGlzIGxhbmd1YWdlLXJvYnVzdC4KIyA9PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PQpDQVRfS08yRU4gPSB7CiAgICAi6rec7KCcIjogIlJlZ3VsYXRpb24iLCAi7J6F67KVIOuPme2WpSI6ICJMZWdpc2xhdGlvbiIsICLtjJDroYAiOiAiTGVnYWwgY2FzZSIsCiAgICAi6riw7IKsIjogIk5ld3MiLCAi7KCV67aAIOuwjyDquLDsl4UgRVNHIOyekOujjCI6ICJHb3YvQ29ycCBFU0ciLAogICAgIuq1reygnCBFU0cg7J6Q66OMIjogIkludGwgRVNHIiwKfQpDVFJZX0tPMkVOID0geyLrsqDtirjrgqgiOiAiVmlldG5hbSIsICLrp5DroIjsnbTsi5zslYQiOiAiTWFsYXlzaWEifQpDT1JQVVNfQ0FURUdPUklFU19FTiA9IFsiUmVndWxhdGlvbiIsICJMZWdhbCBjYXNlIl0gICAjIHJldHJpZXZhbCBjb3JwdXMgKD0gb2xkIOq3nOygnCvtjJDroYApCkNBU0VfQ0FURUdPUllfRU4gPSAiTGVnYWwgY2FzZSIKCmRlZiBub3JtYWxpemVfY2F0X2NvdW50cnkoZGYpOgogICAgIiIiTWFwIEtvcmVhbiBjYXRlZ29yeS9jb3VudHJ5IHRvIEVuZ2xpc2ggaW4tcGxhY2Utc2FmZSAobm8tb3AgaWYgYWxyZWFkeSBFTikuIiIiCiAgICBvdXQgPSBkZi5jb3B5KCkKICAgIGlmICJjYXRlZ29yeSIgaW4gb3V0LmNvbHVtbnM6CiAgICAgICAgb3V0WyJjYXRlZ29yeSJdID0gb3V0WyJjYXRlZ29yeSJdLnJlcGxhY2UoQ0FUX0tPMkVOKQogICAgaWYgImNvdW50cnkiIGluIG91dC5jb2x1bW5zOgogICAgICAgIG91dFsiY291bnRyeSJdID0gb3V0WyJjb3VudHJ5Il0ucmVwbGFjZShDVFJZX0tPMkVOKQogICAgcmV0dXJuIG91dAo=',
    'data_loader': 'IiIiCmRhdGFfbG9hZGVyLnB5IOKAlCBwYXJzZSByYXcvbGFiZWwgemlwcyBpbnRvIERhdGFGcmFtZXMgKGRldGVybWluaXN0aWMpLgoKUjEgRklYIChmaWxlbmFtZSByb2J1c3RuZXNzKQotLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tClJhdyBBSSBIdWIgemlwcyB1c2UgaW5jb25zaXN0ZW50IHNlcGFyYXRvcnMgaW4gdGhlaXIgcmVhbCBmaWxlbmFtZXMsIGUuZy46CiAgICAiVFNfMS4g67Kg7Yq464KoXzAzLiDtjJDroYAuemlwIiAgICAgICJUU18xX+uyoO2KuOuCqF8wMi4g7J6F67KVIOuPme2WpS56aXAiCiAgICAiVFNfMS5f67Kg7Yq464KoXzAxX+q3nOygnC56aXAiICAgICAgICJUU18yLiDrp5DroIjsnbTsi5zslYRfMDYuIOq1reygnCBFU0cg7J6Q66OMLnppcCIKICAgICIwMi7rnbzrsqjrp4HrjbDsnbTthLAuemlwIiAgICAgICAgICAgIChsYWJlbCB6aXApClRoZSBwcmV2aW91cyBsb2FkZXIgcmVsaWVkIG9uIGEgZml4ZWQgU09VUkNFX1pJUFMvTEFCRUxfWklQIG5hbWUgbGlzdCBhbmQgbWF0Y2hlZApub3RoaW5nLiBXZSBub3cgQVVUTy1ESVNDT1ZFUiBkYXRhL3Jhdy8qLnppcCBhbmQgY2xhc3NpZnkgZWFjaCBieSAoY291bnRyeSwKMi1kaWdpdCBjYXRlZ29yeSBjb2RlKSBwYXJzZWQgd2l0aCBhIHNlcGFyYXRvci10b2xlcmFudCByZWdleCwgc2tpcHBpbmcgdGhlCmxhYmVsIHppcC4gRnVuY3Rpb24gc2lnbmF0dXJlcyBhcmUgdW5jaGFuZ2VkLCBzbyBjYWxsZXJzIGFyZSB1bmFmZmVjdGVkLgoKRGVzaWduIChyZXByb2R1Y2liaWxpdHkpOgogIC0gcmVhZCB6aXBzIGZyb20gbWVtb3J5IChubyBleHRyYWN0aW9uKSwKICAtIGl0ZXJhdGUgZW50cmllcyBpbiBzb3J0ZWQgb3JkZXIsCiAgLSByZXR1cm4gRGF0YUZyYW1lcyBzb3J0ZWQgYnkgaWQuCiIiIgpmcm9tIF9fZnV0dXJlX18gaW1wb3J0IGFubm90YXRpb25zCmltcG9ydCBpbwppbXBvcnQganNvbgppbXBvcnQgcmUKaW1wb3J0IHppcGZpbGUKZnJvbSBwYXRobGliIGltcG9ydCBQYXRoCmZyb20gdHlwaW5nIGltcG9ydCBJdGVyYXRvciwgT3B0aW9uYWwKCmltcG9ydCBwYW5kYXMgYXMgcGQKCmZyb20gY29uZmlnIGltcG9ydCBSQVdfRElSCgoKIyA9PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PQojIHJvYnVzdCBmaWxlbmFtZSBjbGFzc2lmaWNhdGlvbiAoc2hhcmVkIGJ5IGFsbCBsb2FkZXJzKQojID09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09CmRlZiBjbGFzc2lmeV9zb3VyY2VfemlwKGZuYW1lKSAtPiBPcHRpb25hbFt0dXBsZVtzdHIsIHN0cl1dOgogICAgIiIiUmV0dXJuIChjb3VudHJ5LCBjYXRjb2RlKSBmb3IgYSBzb3VyY2UgemlwLCBvciBOb25lIGlmIGl0IGlzIG5vdCBvbmUKICAgIChlLmcuIHRoZSBsYWJlbCB6aXApLiBUb2xlcmFudCB0byBzZXBhcmF0b3JzICgnLiAnLCAnXycsICcuXycsICdfXycsICcgJykuIiIiCiAgICBiYXNlID0gUGF0aChmbmFtZSkubmFtZQogICAgaWYgbm90IGJhc2UubG93ZXIoKS5lbmRzd2l0aCgiLnppcCIpOgogICAgICAgIHJldHVybiBOb25lCiAgICBpZiAi652867KoIiBpbiBiYXNlOiAgICAgICAgICAgICAgICAgICAgICAgIyBsYWJlbCB6aXAgLT4gaGFuZGxlZCBzZXBhcmF0ZWx5CiAgICAgICAgcmV0dXJuIE5vbmUKICAgIGlmICLrsqDtirjrgqgiIGluIGJhc2U6CiAgICAgICAgY291bnRyeSA9ICLrsqDtirjrgqgiCiAgICBlbGlmICLrp5DroIjsnbTsi5zslYQiIGluIGJhc2U6CiAgICAgICAgY291bnRyeSA9ICLrp5DroIjsnbTsi5zslYQiCiAgICBlbHNlOgogICAgICAgIHJldHVybiBOb25lCiAgICBtID0gcmUuc2VhcmNoKHIiXyhcZHsyfSlbLl8gXSIsIGJhc2UpICAgICAjIDItZGlnaXQgY2F0ZWdvcnkgY29kZSBhZnRlciBjb3VudHJ5CiAgICBpZiBub3QgbToKICAgICAgICByZXR1cm4gTm9uZQogICAgcmV0dXJuIGNvdW50cnksIG0uZ3JvdXAoMSkKCgpkZWYgZmluZF9zb3VyY2VfemlwcyhyYXdfZGlyOiBQYXRoID0gUkFXX0RJUik6CiAgICAiIiJbKHBhdGgsIGNvdW50cnksIGNhdGNvZGUpLCAuLi5dIGF1dG8tZGlzY292ZXJlZCAmIGRldGVybWluaXN0aWNhbGx5IHNvcnRlZC4iIiIKICAgIGZvdW5kID0gW10KICAgIGZvciBwIGluIHNvcnRlZChQYXRoKHJhd19kaXIpLmdsb2IoIiouemlwIikpOgogICAgICAgIGluZm8gPSBjbGFzc2lmeV9zb3VyY2VfemlwKHAubmFtZSkKICAgICAgICBpZiBpbmZvIGlzIG5vdCBOb25lOgogICAgICAgICAgICBmb3VuZC5hcHBlbmQoKHAsIGluZm9bMF0sIGluZm9bMV0pKQogICAgZm91bmQuc29ydChrZXk9bGFtYmRhIHQ6ICh0WzFdLCB0WzJdLCB0WzBdLm5hbWUpKQogICAgcmV0dXJuIGZvdW5kCgoKZGVmIGZpbmRfbGFiZWxfemlwKHJhd19kaXI6IFBhdGggPSBSQVdfRElSKSAtPiBPcHRpb25hbFtQYXRoXToKICAgICIiIkxvY2F0ZSB0aGUgbGFiZWwgemlwIChjb250YWlucyAn652867KoJykuIiIiCiAgICBmb3IgcCBpbiBzb3J0ZWQoUGF0aChyYXdfZGlyKS5nbG9iKCIqLnppcCIpKToKICAgICAgICBpZiAi652867KoIiBpbiBwLm5hbWU6CiAgICAgICAgICAgIHJldHVybiBwCiAgICByZXR1cm4gTm9uZQoKCiMgPT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT0KIyBzb3VyY2UgZGF0YQojID09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09CmRlZiBfaXRlcl9qc29uX2luX3ppcCh6ZjogemlwZmlsZS5aaXBGaWxlKSAtPiBJdGVyYXRvcltkaWN0XToKICAgIGZvciBuYW1lIGluIHNvcnRlZCh6Zi5uYW1lbGlzdCgpKToKICAgICAgICBpZiBub3QgbmFtZS5lbmRzd2l0aCgiLmpzb24iKToKICAgICAgICAgICAgY29udGludWUKICAgICAgICB3aXRoIHpmLm9wZW4obmFtZSkgYXMgZmg6CiAgICAgICAgICAgIHRyeToKICAgICAgICAgICAgICAgIHlpZWxkIGpzb24ubG9hZChmaCkKICAgICAgICAgICAgZXhjZXB0IGpzb24uSlNPTkRlY29kZUVycm9yOgogICAgICAgICAgICAgICAgY29udGludWUKCgpkZWYgX2ZsYXR0ZW5fbWV0YV9yZWNvcmQoZDogZGljdCkgLT4gZGljdDoKICAgICIiIk9uZSBzb3VyY2UgSlNPTiAtPiBhIGxpZ2h0d2VpZ2h0IG1ldGFkYXRhIHJvdyAoY29udGVudCBleGNsdWRlZCkuIiIiCiAgICBkYXRhID0gZC5nZXQoImRhdGEiLCB7fSkgb3Ige30KICAgIG0gPSBkLmdldCgibWV0YWRhdGEiLCB7fSkgb3Ige30KICAgIHJlZyA9IG0uZ2V0KCJyZWd1bGF0aW9uIiwge30pIG9yIHt9CiAgICBjaSA9IG0uZ2V0KCJjYXNlX2luZm8iLCB7fSkgb3Ige30KICAgIHJlc3VsdCA9IGNpLmdldCgicmVzdWx0Iiwge30pIG9yIHt9CiAgICBjb250ZW50ID0gZGF0YS5nZXQoImNvbnRlbnQiLCB7fSkgb3Ige30KICAgIHJldHVybiB7CiAgICAgICAgImlkIjogICAgICAgICAgICBtLmdldCgiaWQiLCAiIiksCiAgICAgICAgImRvY19pZCI6ICAgICAgICBkYXRhLmdldCgiaWQiLCAiIiksCiAgICAgICAgImZpbGVfbmFtZSI6ICAgICBtLmdldCgiZmlsZV9uYW1lIiwgIiIpLAogICAgICAgICJjYXRlZ29yeSI6ICAgICAgbS5nZXQoImNhdGVnb3J5IiwgIiIpLAogICAgICAgICJjb3VudHJ5IjogICAgICAgbS5nZXQoImNvdW50cnkiLCAiIiksCiAgICAgICAgInJlZ2lvbiI6ICAgICAgICBtLmdldCgicmVnaW9uIiwgIiIpLAogICAgICAgICJ0aXRsZSI6ICAgICAgICAgbS5nZXQoInRpdGxlIiwgIiIpLAogICAgICAgICJhbGlhcyI6ICAgICAgICAgbS5nZXQoImFsaWFzIiwgIiIpLAogICAgICAgICJlc2ciOiAgICAgICAgICAgbS5nZXQoImVzZyIsICIiKSwKICAgICAgICAiZG9jdW1lbnRfdHlwZSI6IG0uZ2V0KCJkb2N1bWVudF90eXBlIiwgIiIpLAogICAgICAgICJwdWJsaXNoZXIiOiAgICAgbS5nZXQoInB1Ymxpc2hlciIsICIiKSwKICAgICAgICAiZW50aXR5IjogICAgICAgIG0uZ2V0KCJlbnRpdHkiLCAiIiksCiAgICAgICAgInB1Ymxpc2hlZF9hdCI6ICBtLmdldCgicHVibGlzaGVkX2F0IiwgIiIpLAogICAgICAgICJzb3VyY2UiOiAgICAgICAgbS5nZXQoInNvdXJjZSIsICIiKSwKICAgICAgICAiY29weXJpZ2h0ZXIiOiAgIG0uZ2V0KCJjb3B5cmlnaHRlciIsICIiKSwKICAgICAgICAibGVuX2tvIjogICAgbGVuKGNvbnRlbnQuZ2V0KCJrbyIsICIiKSBvciAiIiksCiAgICAgICAgImxlbl9lbiI6ICAgIGxlbihjb250ZW50LmdldCgiZW4iLCAiIikgb3IgIiIpLAogICAgICAgICJsZW5fbG9jYWwiOiBsZW4oY29udGVudC5nZXQoImxvY2FsIiwgIiIpIG9yICIiKSwKICAgICAgICAicmVnX25hbWUiOiAgICAgIHJlZy5nZXQoInJlZ3VsYXRpb25fbmFtZSIsICIiKSBvciAiIiwKICAgICAgICAicmVnX3JlZmVyZW5jZSI6IHJlZy5nZXQoInJlZ3VsYXRpb25fcmVmZXJlbmNlIiwgIiIpIG9yICIiLAogICAgICAgICJlZmZlY3RpdmVfZGF0ZSI6IHJlZy5nZXQoImVmZmVjdGl2ZV9kYXRlIiwgIiIpIG9yICIiLAogICAgICAgICJjYXNlX2RldGFpbCI6ICAgICAgIGNpLmdldCgiZGV0YWlsIiwgIiIpIG9yICIiLAogICAgICAgICJwb2xsdXRhbnRfY2F0ZWdvcnkiOiBjaS5nZXQoInBvbGx1dGFudF9jYXRlZ29yeSIsICIiKSBvciAiIiwKICAgICAgICAic2FuY3Rpb25fdHlwZSI6ICAgICByZXN1bHQuZ2V0KCJ0eXBlIiwgIiIpIG9yICIiLAogICAgICAgICJzYW5jdGlvbl9sZXZlbCI6ICAgIHJlc3VsdC5nZXQoImxldmVsIiwgIiIpIG9yICIiLAogICAgICAgICJ2aW9sYXRpb25fZGF0ZSI6ICAgIHJlc3VsdC5nZXQoInZpb2xhdGlvbl9kYXRlIiwgIiIpIG9yICIiLAogICAgICAgICJzYW5jdGlvbl9kYXRlIjogICAgIHJlc3VsdC5nZXQoInNhbmN0aW9uX2RhdGUiLCAiIikgb3IgIiIsCiAgICAgICAgInByZXZpb3VzX3Zpb2xhdGlvbnMiOiBqc29uLmR1bXBzKGNpLmdldCgicHJldmlvdXNfdmlvbGF0aW9ucyIsIFtdKSwgZW5zdXJlX2FzY2lpPUZhbHNlKSwKICAgIH0KCgpkZWYgbG9hZF9zb3VyY2UocmF3X2RpcjogUGF0aCA9IFJBV19ESVIpIC0+IHBkLkRhdGFGcmFtZToKICAgICIiIlBhcnNlIGFsbCBhdXRvLWRpc2NvdmVyZWQgc291cmNlIHppcHMgLT4gbWV0YWRhdGEgRGF0YUZyYW1lIChubyBjb250ZW50KS4iIiIKICAgIHppcHMgPSBmaW5kX3NvdXJjZV96aXBzKHJhd19kaXIpCiAgICBpZiBub3QgemlwczoKICAgICAgICByYWlzZSBGaWxlTm90Rm91bmRFcnJvcigKICAgICAgICAgICAgZiJObyBzb3VyY2UgemlwcyBmb3VuZCB1bmRlciB7cmF3X2Rpcn0uXG4iCiAgICAgICAgICAgIGYiLT4gUHV0IHRoZSBBSSBIdWIgemlwcyBpbiBkYXRhL3Jhdy8gKGFueSBvZiB0aGUgcmVhbCBmaWxlbmFtZSBmb3JtcykuIgogICAgICAgICkKICAgIHJvd3MgPSBbXQogICAgZm9yIHpwYXRoLCBfY291bnRyeSwgX2NjIGluIHppcHM6CiAgICAgICAgd2l0aCB6aXBmaWxlLlppcEZpbGUoenBhdGgpIGFzIHpmOgogICAgICAgICAgICBmb3IgZCBpbiBfaXRlcl9qc29uX2luX3ppcCh6Zik6CiAgICAgICAgICAgICAgICByb3dzLmFwcGVuZChfZmxhdHRlbl9tZXRhX3JlY29yZChkKSkKICAgIGRmID0gcGQuRGF0YUZyYW1lKHJvd3MpLnNvcnRfdmFsdWVzKCJpZCIsIGtpbmQ9Im1lcmdlc29ydCIpLnJlc2V0X2luZGV4KGRyb3A9VHJ1ZSkKICAgIHJldHVybiBkZgoKCmRlZiBsb2FkX2NvbnRlbnQoaWRzOiBzZXQgfCBOb25lID0gTm9uZSwKICAgICAgICAgICAgICAgICBjb3VudHJ5OiBzdHIgfCBOb25lID0gTm9uZSwKICAgICAgICAgICAgICAgICBjYXRlZ29yeV9jb2RlOiBzdHIgfCBOb25lID0gTm9uZSwKICAgICAgICAgICAgICAgICByYXdfZGlyOiBQYXRoID0gUkFXX0RJUikgLT4gcGQuRGF0YUZyYW1lOgogICAgIiIiTG9hZCB0cmktbGluZ3VhbCBjb250ZW50IHNlbGVjdGl2ZWx5IChtZW1vcnktc2FmZSksIGtleWVkIGJ5IG1ldGFkYXRhLmlkLgoKICAgIEF1dG8tZGlzY292ZXJzIGRhdGEvcmF3LyouemlwIChubyBmaXhlZCBuYW1lcykuIE9wdGlvbmFsIGZpbHRlcnM6CiAgICAgIGlkcyAgICAgICAgICAgOiBvbmx5IHRoZXNlIG1ldGFkYXRhIGlkcywKICAgICAgY291bnRyeSAgICAgICA6ICLrsqDtirjrgqgiIC8gIuunkOugiOydtOyLnOyVhCIsCiAgICAgIGNhdGVnb3J5X2NvZGUgOiAiMDEiLi4iMDYiLgogICAgUmV0dXJucyBEYXRhRnJhbWVbaWQsIGNvbnRlbnRfa28sIGNvbnRlbnRfZW4sIGNvbnRlbnRfbG9jYWxdLgogICAgIiIiCiAgICByb3dzID0gW10KICAgIHppcHMgPSBmaW5kX3NvdXJjZV96aXBzKHJhd19kaXIpCiAgICBpZiBub3QgemlwczoKICAgICAgICByYWlzZSBGaWxlTm90Rm91bmRFcnJvcihmIk5vIHNvdXJjZSB6aXBzIGZvdW5kIHVuZGVyIHtyYXdfZGlyfS4iKQogICAgZm9yIHpwYXRoLCB6YywgY2MgaW4gemlwczoKICAgICAgICBpZiBjb3VudHJ5IGFuZCB6YyAhPSBjb3VudHJ5OgogICAgICAgICAgICBjb250aW51ZQogICAgICAgIGlmIGNhdGVnb3J5X2NvZGUgYW5kIGNjICE9IGNhdGVnb3J5X2NvZGU6CiAgICAgICAgICAgIGNvbnRpbnVlCiAgICAgICAgd2l0aCB6aXBmaWxlLlppcEZpbGUoenBhdGgpIGFzIHpmOgogICAgICAgICAgICBmb3IgZCBpbiBfaXRlcl9qc29uX2luX3ppcCh6Zik6CiAgICAgICAgICAgICAgICBfaWQgPSAoZC5nZXQoIm1ldGFkYXRhIiwge30pIG9yIHt9KS5nZXQoImlkIiwgIiIpCiAgICAgICAgICAgICAgICBpZiBpZHMgaXMgbm90IE5vbmUgYW5kIF9pZCBub3QgaW4gaWRzOgogICAgICAgICAgICAgICAgICAgIGNvbnRpbnVlCiAgICAgICAgICAgICAgICBjID0gKGQuZ2V0KCJkYXRhIiwge30pIG9yIHt9KS5nZXQoImNvbnRlbnQiLCB7fSkgb3Ige30KICAgICAgICAgICAgICAgIHJvd3MuYXBwZW5kKHsKICAgICAgICAgICAgICAgICAgICAiaWQiOiBfaWQsCiAgICAgICAgICAgICAgICAgICAgImNvbnRlbnRfa28iOiAgICBjLmdldCgia28iLCAiIikgb3IgIiIsCiAgICAgICAgICAgICAgICAgICAgImNvbnRlbnRfZW4iOiAgICBjLmdldCgiZW4iLCAiIikgb3IgIiIsCiAgICAgICAgICAgICAgICAgICAgImNvbnRlbnRfbG9jYWwiOiBjLmdldCgibG9jYWwiLCAiIikgb3IgIiIsCiAgICAgICAgICAgICAgICB9KQogICAgcmV0dXJuIHBkLkRhdGFGcmFtZShyb3dzKS5zb3J0X3ZhbHVlcygiaWQiLCBraW5kPSJtZXJnZXNvcnQiKS5yZXNldF9pbmRleChkcm9wPVRydWUpCgoKIyA9PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PQojIGxhYmVsIGRhdGEgKG5lc3RlZCB6aXA6IGNvdW50cnkgeCB7UUEsIGJlbmNobWFya30geCBjYXRlZ29yeSkKIyA9PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PQpkZWYgX2l0ZXJfaW5uZXJfbGFiZWxfemlwcyhvdXRlcjogemlwZmlsZS5aaXBGaWxlKSAtPiBJdGVyYXRvclt0dXBsZVtzdHIsIGJ5dGVzXV06CiAgICBmb3IgbmFtZSBpbiBzb3J0ZWQob3V0ZXIubmFtZWxpc3QoKSk6CiAgICAgICAgaWYgbm90IG5hbWUuZW5kc3dpdGgoIi56aXAiKToKICAgICAgICAgICAgY29udGludWUKICAgICAgICB3aXRoIG91dGVyLm9wZW4obmFtZSkgYXMgZmg6CiAgICAgICAgICAgIHlpZWxkIG5hbWUsIGZoLnJlYWQoKQoKCmRlZiBfY2xhc3NpZnlfbGFiZWxfemlwKGlubmVyX25hbWU6IHN0cikgLT4gc3RyOgogICAgYmFzZSA9IFBhdGgoaW5uZXJfbmFtZSkubmFtZQogICAgaWYgIlFBIiBpbiBiYXNlOgogICAgICAgIHJldHVybiAicWEiCiAgICBpZiAi67Kk7LmY66eI7YGsIiBpbiBiYXNlIG9yICJiZW5jaG1hcmsiIGluIGJhc2UubG93ZXIoKToKICAgICAgICByZXR1cm4gImJlbmNobWFyayIKICAgIHJldHVybiAidW5rbm93biIKCgpkZWYgbG9hZF9sYWJlbHMocmF3X2RpcjogUGF0aCA9IFJBV19ESVIpIC0+IHR1cGxlW3BkLkRhdGFGcmFtZSwgcGQuRGF0YUZyYW1lXToKICAgICIiIlBhcnNlIHRoZSBhdXRvLWRpc2NvdmVyZWQgbGFiZWwgemlwIC0+IChxYV9kZiwgYmVuY2htYXJrX2RmKS4iIiIKICAgIHpwYXRoID0gZmluZF9sYWJlbF96aXAocmF3X2RpcikKICAgIGlmIHpwYXRoIGlzIE5vbmU6CiAgICAgICAgcmFpc2UgRmlsZU5vdEZvdW5kRXJyb3IoZiJMYWJlbCB6aXAgKGNvbnRhaW5zICfrnbzrsqgnKSBub3QgZm91bmQgdW5kZXIge3Jhd19kaXJ9LiIpCgogICAgcWFfcm93cywgYm1fcm93cyA9IFtdLCBbXQogICAgd2l0aCB6aXBmaWxlLlppcEZpbGUoenBhdGgpIGFzIG91dGVyOgogICAgICAgIGlubmVyX3ppcHMgPSBsaXN0KF9pdGVyX2lubmVyX2xhYmVsX3ppcHMob3V0ZXIpKQogICAgICAgIGlmIGlubmVyX3ppcHM6CiAgICAgICAgICAgIGl0ZXJhdG9yID0gaW5uZXJfemlwcwogICAgICAgIGVsc2U6CiAgICAgICAgICAgIGl0ZXJhdG9yID0gWyh6cGF0aC5uYW1lLCB6cGF0aC5yZWFkX2J5dGVzKCkpXSAgIyBmbGF0IGxhYmVsIHppcCBmYWxsYmFjawogICAgICAgIGZvciBpbm5lcl9uYW1lLCBpbm5lcl9ieXRlcyBpbiBpdGVyYXRvcjoKICAgICAgICAgICAga2luZCA9IF9jbGFzc2lmeV9sYWJlbF96aXAoaW5uZXJfbmFtZSkKICAgICAgICAgICAgd2l0aCB6aXBmaWxlLlppcEZpbGUoaW8uQnl0ZXNJTyhpbm5lcl9ieXRlcykpIGFzIGlubmVyX3pmOgogICAgICAgICAgICAgICAgZm9yIGQgaW4gX2l0ZXJfanNvbl9pbl96aXAoaW5uZXJfemYpOgogICAgICAgICAgICAgICAgICAgIHFhID0gZC5nZXQoInFhIiwge30pIG9yIHt9CiAgICAgICAgICAgICAgICAgICAgYmFzZSA9IHsKICAgICAgICAgICAgICAgICAgICAgICAgImlkIjogICAgICAgIGQuZ2V0KCJpZCIsICIiKSwKICAgICAgICAgICAgICAgICAgICAgICAgImNhdGVnb3J5IjogIHFhLmdldCgiY2F0ZWdvcnkiLCAiIiksCiAgICAgICAgICAgICAgICAgICAgICAgICJjb3VudHJ5IjogICBxYS5nZXQoImNvdW50cnkiLCAiIiksCiAgICAgICAgICAgICAgICAgICAgICAgICJlc2ciOiAgICAgICBxYS5nZXQoIkVTRyIsICIiKSwKICAgICAgICAgICAgICAgICAgICAgICAgInNvdXJjZV9pZCI6IGpzb24uZHVtcHMocWEuZ2V0KCJzb3VyY2VfaWQiLCBbXSksIGVuc3VyZV9hc2NpaT1GYWxzZSksCiAgICAgICAgICAgICAgICAgICAgfQogICAgICAgICAgICAgICAgICAgIGlmIGtpbmQgPT0gInFhIjoKICAgICAgICAgICAgICAgICAgICAgICAgdHVybnMgPSBxYS5nZXQoInFhX3R1cm5zIiwgW10pCiAgICAgICAgICAgICAgICAgICAgICAgIGJhc2VbIm5fdHVybnMiXSA9IGxlbih0dXJucykKICAgICAgICAgICAgICAgICAgICAgICAgYmFzZVsicWFfdHVybnMiXSA9IGpzb24uZHVtcHModHVybnMsIGVuc3VyZV9hc2NpaT1GYWxzZSkKICAgICAgICAgICAgICAgICAgICAgICAgcWFfcm93cy5hcHBlbmQoYmFzZSkKICAgICAgICAgICAgICAgICAgICBlbGlmIGtpbmQgPT0gImJlbmNobWFyayI6CiAgICAgICAgICAgICAgICAgICAgICAgIHR1cm5zID0gcWEuZ2V0KCJxYV90dXJucyIsIFtdKQogICAgICAgICAgICAgICAgICAgICAgICB0MCA9IHR1cm5zWzBdIGlmIHR1cm5zIGVsc2Uge30KICAgICAgICAgICAgICAgICAgICAgICAgYmFzZVsicXVlc3Rpb24iXSAgICA9IHQwLmdldCgicV9jb250ZW50IiwgIiIpCiAgICAgICAgICAgICAgICAgICAgICAgIGJhc2VbIm9wdGlvbnMiXSAgICAgPSBqc29uLmR1bXBzKHQwLmdldCgiYV9vcHRpb24iLCBbXSksIGVuc3VyZV9hc2NpaT1GYWxzZSkKICAgICAgICAgICAgICAgICAgICAgICAgYmFzZVsiYW5zd2VyX2luZGV4Il0gPSB0MC5nZXQoImFfaW5kZXgiLCAtMSkKICAgICAgICAgICAgICAgICAgICAgICAgYmFzZVsiZXhwbGFuYXRpb24iXSA9IHQwLmdldCgiYV9leHBsYW5hdGlvbiIsICIiKQogICAgICAgICAgICAgICAgICAgICAgICBibV9yb3dzLmFwcGVuZChiYXNlKQoKICAgIHFhX2RmID0gcGQuRGF0YUZyYW1lKHFhX3Jvd3MpLnNvcnRfdmFsdWVzKCJpZCIsIGtpbmQ9Im1lcmdlc29ydCIpLnJlc2V0X2luZGV4KGRyb3A9VHJ1ZSkKICAgIGJtX2RmID0gcGQuRGF0YUZyYW1lKGJtX3Jvd3MpLnNvcnRfdmFsdWVzKCJpZCIsIGtpbmQ9Im1lcmdlc29ydCIpLnJlc2V0X2luZGV4KGRyb3A9VHJ1ZSkKICAgIHJldHVybiBxYV9kZiwgYm1fZGYK',
    'labeling': 'IiIiCmxhYmVsaW5nLnB5IOKAlCDtjJDroYAgY2FzZV9pbmZvIOKGkiDspIDsiJjtjJDsoJUg6rOo65OcIOudvOuyqCDsnKDrj4Qg6rec7LmZCgrquLDtmo3shJwgKGIp7KCIIOq1rO2YhC4g652867Ko7J2AICfqs7Xsi5wg7KSA7IiYJ+qwgCDslYTri4jrnbwgKion67KV7KCBIOyxheyehCDshLHrpr0nKirsnZgg7ZSE66Gd7Iuc7J6E7J2ECuuqheyLnOyggeycvOuhnCDtlZzsoJXtlZzri6Qo6rO864yA7ZW07ISdIOywqOuLqCkuCgrqt5zsuZkg7JqU7KeAOgogIC0gcmVzdWx0KHR5cGUrbGV2ZWwpIO2FjeyKpO2KuOyXkCDsoJzsnqwg7Iug7Zi466eMIOyeiOycvOuptCBOT05fQ09NUExJQU5UCiAgLSDrqbTssYUg7Iug7Zi466eMIOyeiOycvOuptCBDT01QTElBTlQKICAtIOuRmCDri6Qv7Zi87J6sL+u2iOy2qeu2hCDihpIgT1RIRVIgKOyduOyggSDqsoDsiJgpCiAgLSByZXN1bHQg6rO1656AIOKGkiBVTkxBQkVMRUQgKGRldGFpbCDquLDrsJgg7IiY64+ZIOuYkOuKlCDsoJzsmbgpCgrso7zsnZg6IO2ZlO2PkCDri6jsnIQgJ+uPmScoVk5EKeydgCAnMeyLrCDshozshqHruYTsmqkg67aA64u0JyDrk7Eg7KSR66a9IOusuOq1rOyXkOuPhCDtnZTtnogg64KY7JmACiAgICAgIOygnOyerCDsi6DtmLjroZwg7JOw66m0IOqzvOuLpCDtj6ztmo3rkJzri6QuIOuUsOudvOyEnCBTQU5DVElPTl9UT0tFTlPsl5DshJwg7KCc7Jm47ZWc64ukLgoiIiIKZnJvbSBfX2Z1dHVyZV9fIGltcG9ydCBhbm5vdGF0aW9ucwppbXBvcnQgcGFuZGFzIGFzIHBkCgpmcm9tIGNvbmZpZyBpbXBvcnQgKAogICAgU0FOQ1RJT05fVE9LRU5TLCBBQ1FVSVRfVE9LRU5TLAogICAgTEFCRUxfTk9OX0NPTVBMSUFOVCwgTEFCRUxfQ09NUExJQU5ULCBMQUJFTF9PVEhFUiwgTEFCRUxfVU5MQUJFTEVELAopCgoKZGVmIGRlcml2ZV9sYWJlbChzYW5jdGlvbl90eXBlOiBzdHIsIHNhbmN0aW9uX2xldmVsOiBzdHIpIC0+IHN0cjoKICAgICIiIuuLqOydvCDtjJDroYAgcmVzdWx0IOKGkiDspIDsiJjtjJDsoJUg652867KoLiIiIgogICAgcyA9IChzdHIoc2FuY3Rpb25fdHlwZSBvciAiIikgKyAiICIgKyBzdHIoc2FuY3Rpb25fbGV2ZWwgb3IgIiIpKS5zdHJpcCgpCiAgICBpZiBub3QgczoKICAgICAgICByZXR1cm4gTEFCRUxfVU5MQUJFTEVECiAgICBoYXNfc2FuY3Rpb24gPSBhbnkodG9rIGluIHMgZm9yIHRvayBpbiBTQU5DVElPTl9UT0tFTlMpCiAgICBoYXNfYWNxdWl0ID0gYW55KHRvayBpbiBzIGZvciB0b2sgaW4gQUNRVUlUX1RPS0VOUykKICAgIGlmIGhhc19zYW5jdGlvbiBhbmQgbm90IGhhc19hY3F1aXQ6CiAgICAgICAgcmV0dXJuIExBQkVMX05PTl9DT01QTElBTlQKICAgIGlmIGhhc19hY3F1aXQgYW5kIG5vdCBoYXNfc2FuY3Rpb246CiAgICAgICAgcmV0dXJuIExBQkVMX0NPTVBMSUFOVAogICAgcmV0dXJuIExBQkVMX09USEVSCgoKZGVmIGxhYmVsX2Nhc2VzKGNhc2VfZGY6IHBkLkRhdGFGcmFtZSkgLT4gcGQuRGF0YUZyYW1lOgogICAgIiIiCiAgICDtjJDroYAgRGF0YUZyYW1lKGlkLCBzYW5jdGlvbl90eXBlLCBzYW5jdGlvbl9sZXZlbCwgLi4uKeyXkCBsYWJlbCDsl7TsnYQg7LaU6rCA7ZW0IOuwmO2ZmC4KICAgIOyeheugpSDsiJzshJzrpbwg67O07KG07ZWY66mwIOqysOygleuhoOyggS4KICAgICIiIgogICAgb3V0ID0gY2FzZV9kZi5jb3B5KCkKICAgIG91dFsibGFiZWwiXSA9IFsKICAgICAgICBkZXJpdmVfbGFiZWwodCwgbCkKICAgICAgICBmb3IgdCwgbCBpbiB6aXAob3V0WyJzYW5jdGlvbl90eXBlIl0sIG91dFsic2FuY3Rpb25fbGV2ZWwiXSkKICAgIF0KICAgIHJldHVybiBvdXQKCgpkZWYgcmVsaWFiaWxpdHlfc3VtbWFyeShsYWJlbGVkOiBwZC5EYXRhRnJhbWUpIC0+IGRpY3Q6CiAgICAiIiLsnpDrj5kg652867KoIOyLoOuisOuPhCDsmpTslb0o7Iug66KwIOyekOuPmSA9IE5PTitDT01QKS4iIiIKICAgIHZjID0gbGFiZWxlZFsibGFiZWwiXS52YWx1ZV9jb3VudHMoKS50b19kaWN0KCkKICAgIG4gPSBsZW4obGFiZWxlZCkKICAgIHJlbGlhYmxlID0gdmMuZ2V0KExBQkVMX05PTl9DT01QTElBTlQsIDApICsgdmMuZ2V0KExBQkVMX0NPTVBMSUFOVCwgMCkKICAgIHJldHVybiB7CiAgICAgICAgImNvdW50cyI6IHZjLAogICAgICAgICJuIjogbiwKICAgICAgICAicmVsaWFibGVfYXV0byI6IHJlbGlhYmxlLAogICAgICAgICJyZWxpYWJsZV9yYXRpbyI6IHJvdW5kKHJlbGlhYmxlIC8gbiwgNCkgaWYgbiBlbHNlIDAuMCwKICAgIH0KCgojID09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09CiMgUkVWSVNJT04gQURELU9OUyAoUjEpOiBsYWJlbCB2YWxpZGl0eSwgbGVha2FnZSwgYXVkaXQsIHJlY29uY2lsaWF0aW9uIChNNCkKIyA9PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PQpmcm9tIGNvbmZpZyBpbXBvcnQgU0FOQ1RJT05fVE9LRU5TIGFzIF9TQU5DLCBBQ1FVSVRfVE9LRU5TIGFzIF9BQ1EgICMgbm9xYTogRTQwMgoKCmRlZiB2ZXJkaWN0X2xlYWthZ2UoY29udGVudF90ZXh0OiBzdHIpIC0+IGRpY3Q6CiAgICAiIiJEb2VzIHRoZSBjYXNlICpib2R5KiBjb250YWluIGV4cGxpY2l0IHZlcmRpY3Qvc2FuY3Rpb24gd29yZGluZz8KCiAgICBSZXZpZXdlciAyIChNNCk6IGlmIHRoZSBqdWRnZWQgdGV4dCBhbHJlYWR5IHN0YXRlcyB0aGUgb3V0Y29tZSwgdGhlIHRhc2sgbWF5CiAgICByZWR1Y2UgdG8gb3V0Y29tZSBleHRyYWN0aW9uIHJhdGhlciB0aGFuIGNvbXBsaWFuY2UgcmVhc29uaW5nLiBXZSBmbGFnLCBwZXIKICAgIGNhc2UsIHdoZXRoZXIgc2FuY3Rpb24vYWNxdWl0dGFsIHRva2VucyBhcHBlYXIgaW4gdGhlIHRleHQgYWN0dWFsbHkgc2VudCB0bwogICAgdGhlIGp1ZGdlLCBzbyB0aGUgbWFudXNjcmlwdCBjYW4gcmVwb3J0IHRoZSBsZWFrYWdlIHJhdGUgYW5kIChvcHRpb25hbGx5KQogICAgcmUtcnVuIG9uIHZlcmRpY3QtbWFza2VkIHRleHQuCiAgICAiIiIKICAgIHMgPSBzdHIoY29udGVudF90ZXh0IG9yICIiKQogICAgaGl0X3NhbmN0aW9uID0gW3QgZm9yIHQgaW4gX1NBTkMgaWYgdCBpbiBzXQogICAgaGl0X2FjcXVpdCA9IFt0IGZvciB0IGluIF9BQ1EgaWYgdCBpbiBzXQogICAgcmV0dXJuIHsKICAgICAgICAibGVha3NfdmVyZGljdCI6IGJvb2woaGl0X3NhbmN0aW9uIG9yIGhpdF9hY3F1aXQpLAogICAgICAgICJuX3NhbmN0aW9uX3Rva2VucyI6IGxlbihoaXRfc2FuY3Rpb24pLAogICAgICAgICJuX2FjcXVpdF90b2tlbnMiOiBsZW4oaGl0X2FjcXVpdCksCiAgICAgICAgInRva2VucyI6IChoaXRfc2FuY3Rpb24gKyBoaXRfYWNxdWl0KVs6MTBdLAogICAgfQoKCmRlZiBtYXNrX3ZlcmRpY3QoY29udGVudF90ZXh0OiBzdHIsIHBsYWNlaG9sZGVyOiBzdHIgPSAiW1JFREFDVEVEXSIpIC0+IHN0cjoKICAgICIiIlJlcGxhY2Ugc2FuY3Rpb24vYWNxdWl0dGFsIHRva2VucyBzbyB0aGUganVkZ2UgY2Fubm90IGNvcHkgdGhlIG91dGNvbWUuIiIiCiAgICBzID0gc3RyKGNvbnRlbnRfdGV4dCBvciAiIikKICAgIGZvciB0IGluIHNvcnRlZChzZXQoX1NBTkMgKyBfQUNRKSwga2V5PWxlbiwgcmV2ZXJzZT1UcnVlKToKICAgICAgICBzID0gcy5yZXBsYWNlKHQsIHBsYWNlaG9sZGVyKQogICAgcmV0dXJuIHMKCgpkZWYgYXVkaXRfc2FtcGxlKGxhYmVsZWQsIG5fcGVyX2xhYmVsOiBpbnQgPSAzMCwgc2VlZDogaW50ID0gNDIpOgogICAgIiIiRGV0ZXJtaW5pc3RpYyBzdHJhdGlmaWVkIHNhbXBsZSBvZiBhdXRvLWxhYmVsZWQgY2FzZXMgZm9yIGh1bWFuIGF1ZGl0LgoKICAgIFByb2R1Y2VzIGEgcmV2aWV3IHNoZWV0IChpZCwgbGFiZWwsIHNhbmN0aW9uX3R5cGUsIHNhbmN0aW9uX2xldmVsLCBkZXRhaWwpCiAgICB0aGUgYXV0aG9ycyBmaWxsIHdpdGggYSBodW1hbiBqdWRnbWVudCwgc28gaW50ZXItYW5ub3RhdG9yIGFncmVlbWVudCB3aXRoCiAgICB0aGUgcnVsZSBjYW4gYmUgcmVwb3J0ZWQuCiAgICAiIiIKICAgIGltcG9ydCBwYW5kYXMgYXMgcGQKICAgIGNvbHMgPSBbYyBmb3IgYyBpbiBbImlkIiwgImxhYmVsIiwgInNhbmN0aW9uX3R5cGUiLCAic2FuY3Rpb25fbGV2ZWwiLAogICAgICAgICAgICAgICAgICAgICAgICAiY2FzZV9kZXRhaWwiLCAiY291bnRyeSJdIGlmIGMgaW4gbGFiZWxlZC5jb2x1bW5zXQogICAgcGFydHMgPSBbXQogICAgZm9yIGxhYiwgZyBpbiBsYWJlbGVkLmdyb3VwYnkoImxhYmVsIik6CiAgICAgICAgcGFydHMuYXBwZW5kKGcuc2FtcGxlKG49bWluKG5fcGVyX2xhYmVsLCBsZW4oZykpLCByYW5kb21fc3RhdGU9c2VlZClbY29sc10pCiAgICBvdXQgPSBwZC5jb25jYXQocGFydHMpLnNvcnRfdmFsdWVzKFsibGFiZWwiLCAiaWQiXSkucmVzZXRfaW5kZXgoZHJvcD1UcnVlKQogICAgb3V0WyJodW1hbl9sYWJlbCJdID0gIiIgICAgICAgICMgdG8gYmUgZmlsbGVkIGJ5IGFuIGFubm90YXRvcgogICAgb3V0WyJhZ3JlZXNfd2l0aF9ydWxlIl0gPSAiIiAgICMgdG8gYmUgZmlsbGVkIGJ5IGFuIGFubm90YXRvcgogICAgcmV0dXJuIG91dAoKCmRlZiByZWNvbmNpbGVfY291bnRzKHNvdXJjZSkgLT4gZGljdDoKICAgICIiIlJlY29uY2lsZSB0aGUgY2FzZS1jb3VudCBkaXNjcmVwYW5jaWVzIHJldmlld2VyIDIgKG1pbm9yIDIpIGZsYWdnZWQuCgogICAgRXhwbGFpbnM6IHRvdGFsIGNhc2VzIHZzIHBvcHVsYXRlZCBvdXRjb21lcyB2cyBVTkxBQkVMRUQgdnMKICAgIHJlc3VsdHNJblNhbmN0aW9uIGVkZ2VzLCBhbGwgZnJvbSBvbmUgc291cmNlIG9mIHRydXRoLgogICAgIiIiCiAgICBpbXBvcnQgcGFuZGFzIGFzIHBkCiAgICBjYXNlID0gc291cmNlW3NvdXJjZVsiY2F0ZWdvcnkiXS5pc2luKFsiTGVnYWwgY2FzZSIsICLtjJDroYAiXSldLmNvcHkoKQogICAgdG90YWwgPSBsZW4oY2FzZSkKICAgIHJlc3VsdF9maWxsZWQgPSAoY2FzZVsic2FuY3Rpb25fdHlwZSJdLmZpbGxuYSgiIikuYXN0eXBlKHN0cikuc3RyLnN0cmlwKCkgIT0gIiIpLnN1bSgpCiAgICBsYWJlbGVkID0gbGFiZWxfY2FzZXMoY2FzZSkKICAgIHZjID0gbGFiZWxlZFsibGFiZWwiXS52YWx1ZV9jb3VudHMoKS50b19kaWN0KCkKICAgIHJldHVybiB7CiAgICAgICAgInRvdGFsX2Nhc2VzIjogaW50KHRvdGFsKSwKICAgICAgICAicmVzdWx0X2ZpZWxkX2ZpbGxlZCI6IGludChyZXN1bHRfZmlsbGVkKSwKICAgICAgICAicmVzdWx0X2ZpZWxkX2VtcHR5KFVOTEFCRUxFRCkiOiBpbnQodG90YWwgLSByZXN1bHRfZmlsbGVkKSwKICAgICAgICAibGFiZWxfY291bnRzIjoge2s6IGludCh2KSBmb3IgaywgdiBpbiB2Yy5pdGVtcygpfSwKICAgICAgICAibm90ZSI6ICgicmVzdWx0c0luU2FuY3Rpb24gZWRnZXMgY291bnQgb25seSBjYXNlcyB3aG9zZSBydWxlIGxhYmVsIGlzICIKICAgICAgICAgICAgICAgICAiTk9OX0NPTVBMSUFOVDsgT1RIRVIvQ09NUExJQU5UL1VOTEFCRUxFRCBjYXJyeSBubyBzYW5jdGlvbiBlZGdlLiIpLAogICAgfQo=',
    'metrics': 'IiIiCm1ldHJpY3MucHkg4oCUIOq1kOywqOyWuOyWtCDsnbzqtIDshLEg67CPIOykgOyImO2MkOyglSDtj4nqsIAg7KeA7ZGcCgrquLDtmo3shJwgQy0yIOq1rO2YhC4g7Jm467aAIOydmOyhtCDstZzshowobnVtcHnrp4wpLiBMTE0g7JiI7LihIOqysOqzvOqwgCDso7zslrTsp4DrqbQK7Ja47Ja0IOqwhCDsnbzsuZjCt+ygle2ZleuPhCDqsqnssKjCt+uztOygleyYpOywqOulvCDqs4TsgrDtlZzri6QuCiIiIgpmcm9tIF9fZnV0dXJlX18gaW1wb3J0IGFubm90YXRpb25zCmltcG9ydCBudW1weSBhcyBucAoKCmRlZiBhZ3JlZW1lbnQocHJlZF9hOiBsaXN0LCBwcmVkX2I6IGxpc3QpIC0+IGZsb2F0OgogICAgIiIi65GQIOyWuOyWtCDsmIjsuKHsnZgg64uo7IicIOydvOy5mOycqC4iIiIKICAgIGEgPSBucC5hc2FycmF5KHByZWRfYSk7IGIgPSBucC5hc2FycmF5KHByZWRfYikKICAgIGFzc2VydCBsZW4oYSkgPT0gbGVuKGIpCiAgICByZXR1cm4gZmxvYXQoKGEgPT0gYikubWVhbigpKSBpZiBsZW4oYSkgZWxzZSAwLjAKCgpkZWYgY29oZW5fa2FwcGEocHJlZF9hOiBsaXN0LCBwcmVkX2I6IGxpc3QsIGxhYmVsczogbGlzdCB8IE5vbmUgPSBOb25lKSAtPiBmbG9hdDoKICAgICIiIkNvaGVuJ3MgzrogKOyasOyXsCDrs7TsoJUg7J287LmYKS4g65GQIHJhdGVyID0g65GQIOyWuOyWtCDsmIjsuKEuIiIiCiAgICBhID0gbnAuYXNhcnJheShwcmVkX2EpOyBiID0gbnAuYXNhcnJheShwcmVkX2IpCiAgICBpZiBsYWJlbHMgaXMgTm9uZToKICAgICAgICBsYWJlbHMgPSBzb3J0ZWQoc2V0KGEudG9saXN0KCkpIHwgc2V0KGIudG9saXN0KCkpKQogICAgaWR4ID0ge2w6IGkgZm9yIGksIGwgaW4gZW51bWVyYXRlKGxhYmVscyl9CiAgICBuID0gbGVuKGEpCiAgICBpZiBuID09IDA6CiAgICAgICAgcmV0dXJuIDAuMAogICAgSyA9IGxlbihsYWJlbHMpCiAgICBjb25mID0gbnAuemVyb3MoKEssIEspKQogICAgZm9yIHgsIHkgaW4gemlwKGEsIGIpOgogICAgICAgIGNvbmZbaWR4W3hdLCBpZHhbeV1dICs9IDEKICAgIHBvID0gbnAudHJhY2UoY29uZikgLyBuCiAgICByb3cgPSBjb25mLnN1bShheGlzPTEpIC8gbgogICAgY29sID0gY29uZi5zdW0oYXhpcz0wKSAvIG4KICAgIHBlID0gZmxvYXQoKHJvdyAqIGNvbCkuc3VtKCkpCiAgICByZXR1cm4gZmxvYXQoKHBvIC0gcGUpIC8gKDEgLSBwZSkpIGlmICgxIC0gcGUpID4gMWUtMTIgZWxzZSAxLjAKCgpkZWYgYWNjdXJhY3kocHJlZDogbGlzdCwgZ29sZDogbGlzdCkgLT4gZmxvYXQ6CiAgICBwID0gbnAuYXNhcnJheShwcmVkKTsgZyA9IG5wLmFzYXJyYXkoZ29sZCkKICAgIHJldHVybiBmbG9hdCgocCA9PSBnKS5tZWFuKCkpIGlmIGxlbihwKSBlbHNlIDAuMAoKCmRlZiBhY2N1cmFjeV9nYXAocHJlZF9lbjogbGlzdCwgcHJlZF9sb2NhbDogbGlzdCwgZ29sZDogbGlzdCkgLT4gZGljdDoKICAgICIiIkFjYyhlbikgLSBBY2MobG9jYWwpOiBIMiDtlbXsi6wg7KeA7ZGcLiIiIgogICAgYWNjX2VuID0gYWNjdXJhY3kocHJlZF9lbiwgZ29sZCkKICAgIGFjY19sb2NhbCA9IGFjY3VyYWN5KHByZWRfbG9jYWwsIGdvbGQpCiAgICByZXR1cm4gewogICAgICAgICJhY2NfZW4iOiByb3VuZChhY2NfZW4sIDQpLAogICAgICAgICJhY2NfbG9jYWwiOiByb3VuZChhY2NfbG9jYWwsIDQpLAogICAgICAgICJnYXAiOiByb3VuZChhY2NfZW4gLSBhY2NfbG9jYWwsIDQpLAogICAgfQoKCmRlZiBleHBlY3RlZF9jYWxpYnJhdGlvbl9lcnJvcihjb25maWRlbmNlczogbGlzdCwgY29ycmVjdDogbGlzdCwgbl9iaW5zOiBpbnQgPSAxMCkgLT4gZmxvYXQ6CiAgICAiIiJFQ0U6IO2ZleyLoOuPhOyZgCDsi6TsoJwg7KCV64u166Wg7J2YIOq0tOumrCjqs7zsi6Ag7KeE64uoKS4iIiIKICAgIGNvbmYgPSBucC5hc2FycmF5KGNvbmZpZGVuY2VzLCBkdHlwZT1mbG9hdCkKICAgIGNvcnIgPSBucC5hc2FycmF5KGNvcnJlY3QsIGR0eXBlPWZsb2F0KQogICAgaWYgbGVuKGNvbmYpID09IDA6CiAgICAgICAgcmV0dXJuIDAuMAogICAgYmlucyA9IG5wLmxpbnNwYWNlKDAsIDEsIG5fYmlucyArIDEpCiAgICBlY2UgPSAwLjAKICAgIG4gPSBsZW4oY29uZikKICAgIGZvciBpIGluIHJhbmdlKG5fYmlucyk6CiAgICAgICAgbG8sIGhpID0gYmluc1tpXSwgYmluc1tpICsgMV0KICAgICAgICBtYXNrID0gKGNvbmYgPiBsbykgJiAoY29uZiA8PSBoaSkgaWYgaSA+IDAgZWxzZSAoY29uZiA+PSBsbykgJiAoY29uZiA8PSBoaSkKICAgICAgICBpZiBtYXNrLnN1bSgpID09IDA6CiAgICAgICAgICAgIGNvbnRpbnVlCiAgICAgICAgYWNjX2JpbiA9IGNvcnJbbWFza10ubWVhbigpCiAgICAgICAgY29uZl9iaW4gPSBjb25mW21hc2tdLm1lYW4oKQogICAgICAgIGVjZSArPSAobWFzay5zdW0oKSAvIG4pICogYWJzKGFjY19iaW4gLSBjb25mX2JpbikKICAgIHJldHVybiBmbG9hdChlY2UpCgoKZGVmIHJldHJpZXZhbF9vdmVybGFwKHRvcGtfYTogbGlzdCwgdG9wa19iOiBsaXN0KSAtPiBmbG9hdDoKICAgICIiIuyWuOyWtOuzhCB0b3AtayDqsoDsg4nqsrDqs7wgSmFjY2FyZCDspJHssqko6rO87KCcIEIg7J286rSA7ISxKS4iIiIKICAgIHNhLCBzYiA9IHNldCh0b3BrX2EpLCBzZXQodG9wa19iKQogICAgaWYgbm90IHNhIGFuZCBub3Qgc2I6CiAgICAgICAgcmV0dXJuIDEuMAogICAgcmV0dXJuIGxlbihzYSAmIHNiKSAvIGxlbihzYSB8IHNiKQoKCiMgPT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT0KIyBSRVZJU0lPTiBBREQtT05TIChSMSk6IHVuY2VydGFpbnR5LCBlcXVpdmFsZW5jZSB0ZXN0aW5nLCBzZWxlY3RpdmUgcHJlZGljdGlvbgojIC0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tCiMgUmV2aWV3ZXIgMjogTTUgKHJpc2stY292ZXJhZ2UgY3VydmVzLCBFQ0UsIHVuY2VydGFpbnR5IG9uIG1haW4gY29tcGFyaXNvbnM7CiMganVzdGlmeSAibmVnbGlnaWJsZSBnYXAiIHZpYSBhbiBlcXVpdmFsZW5jZSB0ZXN0IHdpdGggYSBzdGF0ZWQgbWFyZ2luKSBhbmQKIyBSZXN1bHRzIFE0IChDSXMgbmVlZGVkOyBhbiBvYnNlcnZlZCAwLjAwOCBkb2VzIG5vdCBlc3RhYmxpc2ggbmVnbGlnaWJpbGl0eSkuCiMgPT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT0KCgpkZWYgYm9vdHN0cmFwX2NpKHZhbHVlcywgc3RhdF9mbj1ucC5tZWFuLCBuX2Jvb3Q6IGludCA9IDEwMDAsCiAgICAgICAgICAgICAgICAgY2k6IGZsb2F0ID0gMC45NSwgc2VlZDogaW50ID0gNDIpIC0+IGRpY3Q6CiAgICAiIiJOb25wYXJhbWV0cmljIGJvb3RzdHJhcCBDSSBmb3IgYSBzdGF0aXN0aWMgb3ZlciBhIDEtRCBzYW1wbGUuIiIiCiAgICBybmcgPSBucC5yYW5kb20uZGVmYXVsdF9ybmcoc2VlZCkKICAgIHggPSBucC5hc2FycmF5KHZhbHVlcywgZHR5cGU9ZmxvYXQpCiAgICBuID0gbGVuKHgpCiAgICBpZiBuID09IDA6CiAgICAgICAgcmV0dXJuIHsicG9pbnQiOiAwLjAsICJsbyI6IDAuMCwgImhpIjogMC4wLCAibiI6IDB9CiAgICBib290cyA9IG5wLmVtcHR5KG5fYm9vdCkKICAgIGZvciBiIGluIHJhbmdlKG5fYm9vdCk6CiAgICAgICAgaWR4ID0gcm5nLmludGVnZXJzKDAsIG4sIG4pCiAgICAgICAgYm9vdHNbYl0gPSBzdGF0X2ZuKHhbaWR4XSkKICAgIGxvID0gZmxvYXQobnAucXVhbnRpbGUoYm9vdHMsICgxIC0gY2kpIC8gMikpCiAgICBoaSA9IGZsb2F0KG5wLnF1YW50aWxlKGJvb3RzLCAxIC0gKDEgLSBjaSkgLyAyKSkKICAgIHJldHVybiB7InBvaW50IjogZmxvYXQoc3RhdF9mbih4KSksICJsbyI6IGxvLCAiaGkiOiBoaSwgIm4iOiBufQoKCmRlZiBwYWlyZWRfZGlmZl9jaShjb3JyZWN0X2EsIGNvcnJlY3RfYiwgbl9ib290OiBpbnQgPSAxMDAwLAogICAgICAgICAgICAgICAgICAgY2k6IGZsb2F0ID0gMC45NSwgc2VlZDogaW50ID0gNDIpIC0+IGRpY3Q6CiAgICAiIiJCb290c3RyYXAgQ0kgZm9yIHRoZSBQQUlSRUQgZGlmZmVyZW5jZSBpbiBhY2N1cmFjeSAoYWNjX2EgLSBhY2NfYikuCgogICAgY29ycmVjdF9hIC8gY29ycmVjdF9iIGFyZSAwLzEgY29ycmVjdG5lc3MgYXJyYXlzIG9uIHRoZSBTQU1FIGl0ZW1zCiAgICAoZS5nLiBFbmdsaXNoIHZzIGxvY2FsIG9uIHRoZSBzYW1lIGNhc2VzKS4gUGFpcmVkIHJlc2FtcGxpbmcga2VlcHMgdGhlCiAgICBkZXBlbmRlbmNlIHN0cnVjdHVyZSwgd2hpY2ggYW4gdW5wYWlyZWQgQ0kgd291bGQgaWdub3JlLgogICAgIiIiCiAgICBybmcgPSBucC5yYW5kb20uZGVmYXVsdF9ybmcoc2VlZCkKICAgIGEgPSBucC5hc2FycmF5KGNvcnJlY3RfYSwgZHR5cGU9ZmxvYXQpCiAgICBiID0gbnAuYXNhcnJheShjb3JyZWN0X2IsIGR0eXBlPWZsb2F0KQogICAgYXNzZXJ0IGxlbihhKSA9PSBsZW4oYiksICJwYWlyZWQgYXJyYXlzIG11c3QgYmUgZXF1YWwgbGVuZ3RoIgogICAgbiA9IGxlbihhKQogICAgaWYgbiA9PSAwOgogICAgICAgIHJldHVybiB7ImRpZmYiOiAwLjAsICJsbyI6IDAuMCwgImhpIjogMC4wLCAibiI6IDB9CiAgICBib290cyA9IG5wLmVtcHR5KG5fYm9vdCkKICAgIGZvciBpIGluIHJhbmdlKG5fYm9vdCk6CiAgICAgICAgaWR4ID0gcm5nLmludGVnZXJzKDAsIG4sIG4pCiAgICAgICAgYm9vdHNbaV0gPSBhW2lkeF0ubWVhbigpIC0gYltpZHhdLm1lYW4oKQogICAgbG8gPSBmbG9hdChucC5xdWFudGlsZShib290cywgKDEgLSBjaSkgLyAyKSkKICAgIGhpID0gZmxvYXQobnAucXVhbnRpbGUoYm9vdHMsIDEgLSAoMSAtIGNpKSAvIDIpKQogICAgcmV0dXJuIHsiZGlmZiI6IGZsb2F0KGEubWVhbigpIC0gYi5tZWFuKCkpLCAibG8iOiBsbywgImhpIjogaGksICJuIjogbn0KCgpkZWYgdG9zdF9lcXVpdmFsZW5jZShjb3JyZWN0X2EsIGNvcnJlY3RfYiwgbWFyZ2luOiBmbG9hdCA9IDAuMDUsCiAgICAgICAgICAgICAgICAgICAgIG5fYm9vdDogaW50ID0gMjAwMCwgc2VlZDogaW50ID0gNDIpIC0+IGRpY3Q6CiAgICAiIiJUd28gT25lLVNpZGVkIFRlc3RzIChUT1NUKSBmb3IgZXF1aXZhbGVuY2Ugb2YgcGFpcmVkIGFjY3VyYWNpZXMuCgogICAgSDAgKG5vbi1lcXVpdmFsZW5jZSk6IHxhY2NfYSAtIGFjY19ifCA+PSBtYXJnaW4uCiAgICBXZSByZWplY3QgSDAgKGkuZS4gY29uY2x1ZGUgRVFVSVZBTEVOQ0Ugd2l0aGluICsvLSBtYXJnaW4pIHdoZW4gdGhlCiAgICAoMS0yKmFscGhhKSBib290c3RyYXAgQ0kgb2YgdGhlIHBhaXJlZCBkaWZmZXJlbmNlIGxpZXMgZW50aXJlbHkgaW5zaWRlCiAgICAoLW1hcmdpbiwgK21hcmdpbikuIFJldHVybnMgdGhlIDkwJSBDSSAoYWxwaGE9MC4wNSBlYWNoIHNpZGUpIGFuZCBhIHZlcmRpY3QuCgogICAgVGhpcyBpcyB3aGF0IGxldHMgdGhlIHBhcGVyIHNheSAibmVnbGlnaWJsZSBnYXAiIGRlZmVuc2libHksIGluc3RlYWQgb2YKICAgIGluZmVycmluZyBpdCBmcm9tIGEgc2luZ2xlIHBvaW50IGVzdGltYXRlIG9mIDAuMDA4LgogICAgIiIiCiAgICByZXMgPSBwYWlyZWRfZGlmZl9jaShjb3JyZWN0X2EsIGNvcnJlY3RfYiwgbl9ib290PW5fYm9vdCwgY2k9MC45MCwgc2VlZD1zZWVkKQogICAgZXF1aXZhbGVudCA9IChyZXNbImxvIl0gPiAtbWFyZ2luKSBhbmQgKHJlc1siaGkiXSA8IG1hcmdpbikKICAgIHJldHVybiB7ImRpZmYiOiByZXNbImRpZmYiXSwgImNpOTBfbG8iOiByZXNbImxvIl0sICJjaTkwX2hpIjogcmVzWyJoaSJdLAogICAgICAgICAgICAibWFyZ2luIjogbWFyZ2luLCAiZXF1aXZhbGVudCI6IGJvb2woZXF1aXZhbGVudCksICJuIjogcmVzWyJuIl19CgoKZGVmIHJpc2tfY292ZXJhZ2VfY3VydmUoY29uZmlkZW5jZXMsIGNvcnJlY3QpOgogICAgIiIiU2VsZWN0aXZlLXByZWRpY3Rpb24gcmlzay1jb3ZlcmFnZSBjdXJ2ZS4KCiAgICBTb3J0IGl0ZW1zIGJ5IERFU0NFTkRJTkcgY29uZmlkZW5jZTsgYXQgZWFjaCBwcmVmaXggKGNvdmVyYWdlID0gZnJhY3Rpb24KICAgIGFuc3dlcmVkKSBjb21wdXRlIHJpc2sgPSAxIC0gYWNjdXJhY3kgb24gdGhlIGFuc3dlcmVkIHN1YnNldC4KICAgIFJldHVybnMgYXJyYXlzIChjb3ZlcmFnZSwgcmlzaykgcGx1cyBBVVJDIChhcmVhIHVuZGVyIHJpc2stY292ZXJhZ2UpLgogICAgIiIiCiAgICBjb25mID0gbnAuYXNhcnJheShjb25maWRlbmNlcywgZHR5cGU9ZmxvYXQpCiAgICBjb3JyID0gbnAuYXNhcnJheShjb3JyZWN0LCBkdHlwZT1mbG9hdCkKICAgIG4gPSBsZW4oY29uZikKICAgIGlmIG4gPT0gMDoKICAgICAgICByZXR1cm4geyJjb3ZlcmFnZSI6IG5wLmFycmF5KFtdKSwgInJpc2siOiBucC5hcnJheShbXSksICJhdXJjIjogMC4wfQogICAgb3JkZXIgPSBucC5hcmdzb3J0KC1jb25mKQogICAgY29ycl9zb3J0ZWQgPSBjb3JyW29yZGVyXQogICAgY3VtX2NvcnJlY3QgPSBucC5jdW1zdW0oY29ycl9zb3J0ZWQpCiAgICBrID0gbnAuYXJhbmdlKDEsIG4gKyAxKQogICAgY292ZXJhZ2UgPSBrIC8gbgogICAgcmlzayA9IDEuMCAtIGN1bV9jb3JyZWN0IC8gawogICAgX3RyYXB6ID0gZ2V0YXR0cihucCwgInRyYXBlem9pZCIsIGdldGF0dHIobnAsICJ0cmFweiIsIE5vbmUpKQogICAgYXVyYyA9IGZsb2F0KF90cmFweihyaXNrLCBjb3ZlcmFnZSkpCiAgICByZXR1cm4geyJjb3ZlcmFnZSI6IGNvdmVyYWdlLCAicmlzayI6IHJpc2ssICJhdXJjIjogYXVyY30KCgpkZWYgY292ZXJhZ2VfYXRfcmlzayhjb25maWRlbmNlcywgY29ycmVjdCwgdGFyZ2V0X3Jpc2s6IGZsb2F0KSAtPiBmbG9hdDoKICAgICIiIk1heGltdW0gY292ZXJhZ2UgYWNoaWV2YWJsZSB3aGlsZSBrZWVwaW5nIHNlbGVjdGl2ZSByaXNrIDw9IHRhcmdldC4iIiIKICAgIHJjID0gcmlza19jb3ZlcmFnZV9jdXJ2ZShjb25maWRlbmNlcywgY29ycmVjdCkKICAgIGNvdiwgcmlzayA9IHJjWyJjb3ZlcmFnZSJdLCByY1sicmlzayJdCiAgICBvayA9IGNvdltyaXNrIDw9IHRhcmdldF9yaXNrXQogICAgcmV0dXJuIGZsb2F0KG9rLm1heCgpKSBpZiBsZW4ob2spIGVsc2UgMC4wCgoKZGVmIHNlbGVjdGl2ZV9hY2N1cmFjeShwcmVkcywgZ29sZCwgY29uZmlkZW5jZXMsIHRocmVzaG9sZDogZmxvYXQpOgogICAgIiIiQWNjdXJhY3kgb24gaXRlbXMgd2hvc2UgY29uZmlkZW5jZSA+PSB0aHJlc2hvbGQsIHBsdXMgY292ZXJhZ2UuIiIiCiAgICBwID0gbnAuYXNhcnJheShwcmVkcykKICAgIGcgPSBucC5hc2FycmF5KGdvbGQpCiAgICBjID0gbnAuYXNhcnJheShjb25maWRlbmNlcywgZHR5cGU9ZmxvYXQpCiAgICBtYXNrID0gYyA+PSB0aHJlc2hvbGQKICAgIGNvdiA9IGZsb2F0KG1hc2subWVhbigpKSBpZiBsZW4oYykgZWxzZSAwLjAKICAgIGFjYyA9IGZsb2F0KChwW21hc2tdID09IGdbbWFza10pLm1lYW4oKSkgaWYgbWFzay5zdW0oKSBlbHNlIDAuMAogICAgcmV0dXJuIHsidGhyZXNob2xkIjogdGhyZXNob2xkLCAiY292ZXJhZ2UiOiBjb3YsICJzZWxlY3RpdmVfYWNjIjogYWNjLAogICAgICAgICAgICAibl9hbnN3ZXJlZCI6IGludChtYXNrLnN1bSgpKX0K',
    'ontology': 'IiIiCm9udG9sb2d5LnB5IOKAlCBRdWFsaWZpZWQgT250b2xvZ3kgQW5jaG9yaW5nIChRT0EpOiB0cmlwbGUgLT4gcXVhZCAtPiBwZW50dXBsZS4KCk1vdGl2YXRpb24gKHJldmlld2VyIDIsIE0zIC8gb3JpZ2luYWxpdHkpCi0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tClRoZSBzdWJtaXR0ZWQgIm9udG9sb2d5LWFuY2hvcmVkIiByYW5rZXIgKEIzKSBvbmx5IGFkZGVkIGEgc2NvcmUgYm9udXMgZm9yCm1hdGNoaW5nIHR3byBtZXRhZGF0YSBmaWVsZHMgKGp1cmlzZGljdGlvbiArIEVTRyB0b3BpYykuIEEgcmV2aWV3ZXIgcmlnaHRseQpub3RlZCB0aGlzIGlzIGEgKm1ldGFkYXRhIGJvb3N0Kiwgbm90IGEgbWV0aG9kb2xvZ2ljYWwgY29udHJpYnV0aW9uLCBhbmQgdGhhdAphIGNvbnZlbnRpb25hbCBtZXRhZGF0YSBiYXNlbGluZSBtdXN0IGJlIGNvbXBhcmVkIGFnYWluc3QuCgpXZSBrZWVwIHRoYXQgZXhhY3QgQjMgYXMgdGhlICoqdHJpcGxlKiogbGV2ZWwgKHRoZSByZXF1aXJlZCBtZXRhZGF0YSBiYXNlbGluZSkKYW5kIGV4dGVuZCB0aGUgb250b2xvZ3kgZnJvbSBiaW5hcnkgdHJpcGxlcyAocywgcCwgbykgdG8gKipuLWFyeSBxdWFsaWZpZWQKc3RhdGVtZW50cyoqIHRoYXQgZW5jb2RlIHRoZSByZWFsIGNvbXBsaWFuY2Ugc2VtYW50aWNzIHRoZSBmbGF0IHRyaXBsZSBjYW5ub3Q6CgogICogdHJpcGxlICAgKHMsIHAsIG8pICAgICAgICAgICAgICAgICAgICAgICAgIGUuZy4gKENhc2UsIGdvdmVybmVkQnksIFJlZ3VsYXRpb24pCiAgKiBxdWFkICAgICAocywgcCwgbyB8IHZhbGlkVGltZSkgICAgICAgICAgICAgKyB0aGUgcmVndWxhdGlvbiBtdXN0IGJlIElOIEZPUkNFCiAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICBhdCB0aGUgdmlvbGF0aW9uL3B1YmxpY2F0aW9uIHRpbWUKICAqIHBlbnR1cGxlIChzLCBwLCBvIHwgdmFsaWRUaW1lLCBzZXZlcml0eSkgICArIHRoZSBzYW5jdGlvbiBzZXZlcml0eSAvIEVTRwogICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgc3VidG9waWMgbXVzdCBiZSBjb25zaXN0ZW50CgpFYWNoIHN0YXRlbWVudCBpcyByZWlmaWVkIChSREYtc3RhciBzdHlsZSkgaW50byBhIHN0YXRlbWVudCBub2RlIGNhcnJ5aW5nCnF1YWxpZmllcnMsIHNvIHRoZSBncmFwaCBpcyBhIHRydWUgbi1hcnkga25vd2xlZGdlIGdyYXBoIHJhdGhlciB0aGFuIGEgYmFnIG9mCnRyaXBsZXMuIFRoZSByZXRyaWV2ZXIgKHNyYy9yZXRyaWV2YWwucHkpIHNjb3JlcyBhIGNhbmRpZGF0ZSBieSBhIHdlaWdodGVkIHN1bQpvZiBwZXItcXVhbGlmaWVyIGFsaWdubWVudCwgd2l0aCB0aGUgd2VpZ2h0IHZlY3RvciB0dW5lZCBvbiBhIGhlbGQtb3V0IHNwbGl0LgpUaGlzIHlpZWxkcyBhbiBhYmxhdGFibGUgbWV0aG9kICh0cmlwbGUgLT4gcXVhZCAtPiBwZW50dXBsZSkgd2hvc2UgZ2FpbiBvdmVyCnRoZSBtZXRhZGF0YSBiYXNlbGluZSBpcyBhdHRyaWJ1dGFibGUgdG8gdGhlICpxdWFsaWZpZXJzKiwgbm90IHRvIHRoZSBmYWN0IHRoYXQKYW4gb250b2xvZ3kgZXhpc3RzLgoKRXZlcnl0aGluZyBoZXJlIGlzIGRlcml2ZWQgMToxIGZyb20gc291cmNlIGZpZWxkcyAoZGVmZW5zaWJsZSwgbm8gaGFuZCBjdXJhdGlvbikKYW5kIGlzIGRldGVybWluaXN0aWMuCiIiIgpmcm9tIF9fZnV0dXJlX18gaW1wb3J0IGFubm90YXRpb25zCgppbXBvcnQgcmUKZnJvbSB0eXBpbmcgaW1wb3J0IE9wdGlvbmFsCgppbXBvcnQgbnVtcHkgYXMgbnAKaW1wb3J0IHBhbmRhcyBhcyBwZAoKZnJvbSBjb25maWcgaW1wb3J0ICgKICAgIFNFVkVSSVRZX01BUCwgU0VWRVJJVFlfT1JERVIsIFFPQV9URU1QT1JBTF9UQVUsCiAgICBTQU5DVElPTl9UT0tFTlMsIEFDUVVJVF9UT0tFTlMsCikKCiMgLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0KIyAwLiBmaWVsZCBub3JtYWxpc2F0aW9uIGhlbHBlcnMKIyAtLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLQpfWUVBUl9SRSA9IHJlLmNvbXBpbGUociIoMTl8MjApXGR7Mn0iKQoKCmRlZiBleHRyYWN0X3llYXIoKnZhbHVlcykgLT4gT3B0aW9uYWxbaW50XToKICAgICIiIkZpcnN0IDQtZGlnaXQgeWVhciBmb3VuZCBhY3Jvc3MgdGhlIGdpdmVuIGRhdGUtaXNoIHN0cmluZ3MsIGVsc2UgTm9uZS4iIiIKICAgIGZvciB2IGluIHZhbHVlczoKICAgICAgICBpZiB2IGlzIE5vbmU6CiAgICAgICAgICAgIGNvbnRpbnVlCiAgICAgICAgbSA9IF9ZRUFSX1JFLnNlYXJjaChzdHIodikpCiAgICAgICAgaWYgbToKICAgICAgICAgICAgeSA9IGludChtLmdyb3VwKDApKQogICAgICAgICAgICBpZiAxOTUwIDw9IHkgPD0gMjEwMDoKICAgICAgICAgICAgICAgIHJldHVybiB5CiAgICByZXR1cm4gTm9uZQoKCmRlZiBzZXZlcml0eV9idWNrZXQoc2FuY3Rpb25fdHlwZTogc3RyLCBzYW5jdGlvbl9sZXZlbDogc3RyKSAtPiBzdHI6CiAgICAiIiJNYXAgYSBjYXNlIHJlc3VsdCB0byBhbiBvcmRpbmFsIHNldmVyaXR5IGJ1Y2tldCAocGVudHVwbGUgcXVhbGlmaWVyKS4iIiIKICAgIHMgPSAoc3RyKHNhbmN0aW9uX3R5cGUgb3IgIiIpICsgIiAiICsgc3RyKHNhbmN0aW9uX2xldmVsIG9yICIiKSkKICAgIGlmIGFueSh0b2sgaW4gcyBmb3IgdG9rIGluIEFDUVVJVF9UT0tFTlMpIGFuZCBub3QgYW55KHRvayBpbiBzIGZvciB0b2sgaW4gU0FOQ1RJT05fVE9LRU5TKToKICAgICAgICByZXR1cm4gIk5PTkUiCiAgICBiZXN0ID0gIk5PTkUiCiAgICBmb3IgdG9rLCBidWNrZXQgaW4gU0VWRVJJVFlfTUFQLml0ZW1zKCk6CiAgICAgICAgaWYgdG9rIGluIHMgYW5kIFNFVkVSSVRZX09SREVSLmluZGV4KGJ1Y2tldCkgPiBTRVZFUklUWV9PUkRFUi5pbmRleChiZXN0KToKICAgICAgICAgICAgYmVzdCA9IGJ1Y2tldAogICAgIyBhIGdlbmVyaWMgc2FuY3Rpb24gdG9rZW4gd2l0aCBubyBzZXZlcml0eSBtYXAgc3RpbGwgY291bnRzIGFzIGF0IGxlYXN0IE1JTk9SCiAgICBpZiBiZXN0ID09ICJOT05FIiBhbmQgYW55KHRvayBpbiBzIGZvciB0b2sgaW4gU0FOQ1RJT05fVE9LRU5TKToKICAgICAgICBiZXN0ID0gIk1JTk9SIgogICAgcmV0dXJuIGJlc3QKCgpkZWYgZXNnX3N1YnRvcGljKHBvbGx1dGFudF9jYXRlZ29yeTogc3RyLCBlc2c6IHN0cikgLT4gc3RyOgogICAgIiIiQSBmaW5lciB0b3BpYyBxdWFsaWZpZXI6IHBvbGx1dGFudF9jYXRlZ29yeSB3aGVuIHByZXNlbnQsIGVsc2UgRVNHIHR5cGUuIiIiCiAgICBwYyA9IHN0cihwb2xsdXRhbnRfY2F0ZWdvcnkgb3IgIiIpLnN0cmlwKCkKICAgIHJldHVybiBwYyBpZiBwYyBlbHNlIHN0cihlc2cgb3IgIiIpLnN0cmlwKCkKCgojIC0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tCiMgMS4gZG9jdW1lbnQgYW5jaG9yIHByb2ZpbGVzICAob25lIHJvdyBwZXIgZG9jdW1lbnQpCiMgLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0KQU5DSE9SX0NPTFMgPSBbImlkIiwgImRvY19pZCIsICJjb3VudHJ5IiwgInJlZ2lvbiIsICJlc2ciLCAic3VidG9waWMiLAogICAgICAgICAgICAgICAieWVhciIsICJkb2N0eXBlIiwgInNldmVyaXR5IiwgImhhc19yZWdfcmVmIl0KCgpkZWYgYnVpbGRfYW5jaG9yX3Byb2ZpbGVzKHNvdXJjZTogcGQuRGF0YUZyYW1lKSAtPiBwZC5EYXRhRnJhbWU6CiAgICAiIiJEZXJpdmUgdGhlIHBlci1kb2N1bWVudCBxdWFsaWZpZWQgYW5jaG9yIHByb2ZpbGUgZnJvbSBzb3VyY2UgbWV0YWRhdGEuCgogICAgUGFyYW1ldGVycwogICAgLS0tLS0tLS0tLQogICAgc291cmNlIDogdGhlIGludGVyaW0gc291cmNlIERhdGFGcmFtZSBmcm9tIG5iMDEgKGRhdGEvaW50ZXJpbS9zb3VyY2UucGFycXVldCkKCiAgICBSZXR1cm5zCiAgICAtLS0tLS0tCiAgICBEYXRhRnJhbWVbQU5DSE9SX0NPTFNdCiAgICAiIiIKICAgIGRmID0gc291cmNlLmNvcHkoKQogICAgeWVhcnMgPSBbCiAgICAgICAgZXh0cmFjdF95ZWFyKGEsIGIsIGMpCiAgICAgICAgZm9yIGEsIGIsIGMgaW4gemlwKGRmLmdldCgiZWZmZWN0aXZlX2RhdGUiLCAiIiksCiAgICAgICAgICAgICAgICAgICAgICAgICAgIGRmLmdldCgidmlvbGF0aW9uX2RhdGUiLCAiIiksCiAgICAgICAgICAgICAgICAgICAgICAgICAgIGRmLmdldCgicHVibGlzaGVkX2F0IiwgIiIpKQogICAgXQogICAgb3V0ID0gcGQuRGF0YUZyYW1lKHsKICAgICAgICAiaWQiOiAgICAgICAgZGZbImlkIl0sCiAgICAgICAgImRvY19pZCI6ICAgIGRmLmdldCgiZG9jX2lkIiwgZGZbImlkIl0pLAogICAgICAgICJjb3VudHJ5IjogICBkZlsiY291bnRyeSJdLmZpbGxuYSgiIikuYXN0eXBlKHN0ciksCiAgICAgICAgInJlZ2lvbiI6ICAgIGRmLmdldCgicmVnaW9uIiwgIiIpLmZpbGxuYSgiIikuYXN0eXBlKHN0ciksCiAgICAgICAgImVzZyI6ICAgICAgIGRmWyJlc2ciXS5maWxsbmEoIiIpLmFzdHlwZShzdHIpLAogICAgICAgICJzdWJ0b3BpYyI6ICBbZXNnX3N1YnRvcGljKHBjLCBlKQogICAgICAgICAgICAgICAgICAgICAgZm9yIHBjLCBlIGluIHppcChkZi5nZXQoInBvbGx1dGFudF9jYXRlZ29yeSIsICIiKSwgZGYuZ2V0KCJlc2ciLCAiIikpXSwKICAgICAgICAieWVhciI6ICAgICAgeWVhcnMsCiAgICAgICAgImRvY3R5cGUiOiAgIGRmLmdldCgiZG9jdW1lbnRfdHlwZSIsIGRmLmdldCgiY2F0ZWdvcnkiLCAiIikpLmZpbGxuYSgiIikuYXN0eXBlKHN0ciksCiAgICAgICAgInNldmVyaXR5IjogIFtzZXZlcml0eV9idWNrZXQodCwgbCkKICAgICAgICAgICAgICAgICAgICAgIGZvciB0LCBsIGluIHppcChkZi5nZXQoInNhbmN0aW9uX3R5cGUiLCAiIiksIGRmLmdldCgic2FuY3Rpb25fbGV2ZWwiLCAiIikpXSwKICAgICAgICAiaGFzX3JlZ19yZWYiOiAoZGYuZ2V0KCJyZWdfcmVmZXJlbmNlIiwgIiIpLmZpbGxuYSgiIikuYXN0eXBlKHN0cikuc3RyLnN0cmlwKCkgIT0gIiIpLAogICAgfSkKICAgIHJldHVybiBvdXRbQU5DSE9SX0NPTFNdLnJlc2V0X2luZGV4KGRyb3A9VHJ1ZSkKCgojIC0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tCiMgMi4gcmVpZmllZCBxdWFsaWZpZWQgc3RhdGVtZW50cyAgKHRoZSBuLWFyeSBLRyBpdHNlbGYpCiMgLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0KZGVmIGJ1aWxkX3F1YWxpZmllZF9zdGF0ZW1lbnRzKHNvdXJjZTogcGQuRGF0YUZyYW1lKSAtPiBwZC5EYXRhRnJhbWU6CiAgICAiIiJFbWl0IHJlaWZpZWQgc3RhdGVtZW50cyB3aXRoIHF1YWxpZmllcnMuCgogICAgV2UgbWF0ZXJpYWxpc2UgdGhlIGNvbXBsaWFuY2UtcmVsZXZhbnQgcmVsYXRpb24gYGdvdmVybmVkQnlgCiAgICAoQ2FzZS9Eb2N1bWVudCAtPiBKdXJpc2RpY3Rpb24rVG9waWMgY29udGV4dCkgYXMgYSBzdGF0ZW1lbnQgbm9kZSBjYXJyeWluZwogICAgdGhlIHZhbGlkVGltZSAoeWVhcikgYW5kIHNldmVyaXR5IHF1YWxpZmllcnMuIFRoZSBwb2ludCBpcyBub3QgdGhlIHNwZWNpZmljCiAgICB0cmlwbGUgYnV0IHRoYXQgZWFjaCBzdGF0ZW1lbnQgaXMgKm4tYXJ5KjogaXQgaXMgKHMscCxvKSBQTFVTIHF1YWxpZmllcnMsCiAgICB3aGljaCB0aGUgdHJpcGxlIHN0b3JlIGNhbm5vdCByZXByZXNlbnQgd2l0aG91dCByZWlmaWNhdGlvbi4KCiAgICBSZXR1cm5zCiAgICAtLS0tLS0tCiAgICBEYXRhRnJhbWVbc3RtdF9pZCwgc3ViamVjdCwgcHJlZGljYXRlLCBvYmplY3QsIHFfanVyaXNkaWN0aW9uLCBxX3RvcGljLAogICAgICAgICAgICAgIHFfdmFsaWR0aW1lLCBxX3NldmVyaXR5LCBxX2RvY3R5cGUsIGFyaXR5XQogICAgIiIiCiAgICBhbmMgPSBidWlsZF9hbmNob3JfcHJvZmlsZXMoc291cmNlKQogICAgcm93cyA9IFtdCiAgICBmb3IgaSwgciBpbiBhbmMuaXRlcnJvd3MoKToKICAgICAgICBqdXJpcyA9IGYie3JbJ2NvdW50cnknXX0ve3JbJ3JlZ2lvbiddfSIuc3RyaXAoIi8iKQogICAgICAgIHJvd3MuYXBwZW5kKHsKICAgICAgICAgICAgInN0bXRfaWQiOiAgIGYiU3tpOjA3ZH0iLAogICAgICAgICAgICAic3ViamVjdCI6ICAgclsiaWQiXSwKICAgICAgICAgICAgInByZWRpY2F0ZSI6ICJoYXNDb21wbGlhbmNlQ29udGV4dCIsCiAgICAgICAgICAgICJvYmplY3QiOiAgICBmIkNUWDo6e2p1cmlzfTo6e3JbJ2VzZyddfSIsCiAgICAgICAgICAgICJxX2p1cmlzZGljdGlvbiI6IGp1cmlzLAogICAgICAgICAgICAicV90b3BpYyI6ICAgICAgICByWyJzdWJ0b3BpYyJdIG9yIHJbImVzZyJdLAogICAgICAgICAgICAicV92YWxpZHRpbWUiOiAgICByWyJ5ZWFyIl0sCiAgICAgICAgICAgICJxX3NldmVyaXR5IjogICAgIHJbInNldmVyaXR5Il0sCiAgICAgICAgICAgICJxX2RvY3R5cGUiOiAgICAgIHJbImRvY3R5cGUiXSwKICAgICAgICB9KQogICAgc3RtdHMgPSBwZC5EYXRhRnJhbWUocm93cykKICAgICMgYXJpdHkgPSAzICh0cmlwbGUpICsgbnVtYmVyIG9mIG5vbi1lbXB0eSBxdWFsaWZpZXJzIHByZXNlbnQKICAgIHFjb2xzID0gWyJxX2p1cmlzZGljdGlvbiIsICJxX3RvcGljIiwgInFfdmFsaWR0aW1lIiwgInFfc2V2ZXJpdHkiLCAicV9kb2N0eXBlIl0KCiAgICBkZWYgX2FyaXR5KHJvdyk6CiAgICAgICAgcHJlc2VudCA9IHN1bSgxIGZvciBjIGluIHFjb2xzIGlmIChyb3dbY10gbm90IGluICgiIiwgTm9uZSkgYW5kIG5vdCAoaXNpbnN0YW5jZShyb3dbY10sIGZsb2F0KSBhbmQgbnAuaXNuYW4ocm93W2NdKSkpKQogICAgICAgIHJldHVybiAzICsgcHJlc2VudAogICAgc3RtdHNbImFyaXR5Il0gPSBzdG10cy5hcHBseShfYXJpdHksIGF4aXM9MSkKICAgIHJldHVybiBzdG10cwoKCmRlZiBzdGF0ZW1lbnRfc3VtbWFyeShzdG10czogcGQuRGF0YUZyYW1lKSAtPiBwZC5EYXRhRnJhbWU6CiAgICAiIiJDb3VudHMgb2Ygc3RhdGVtZW50cyBieSBlZmZlY3RpdmUgYXJpdHkgKHRyaXBsZS9xdWFkL3BlbnR1cGxlKykuIiIiCiAgICB2YyA9IHN0bXRzWyJhcml0eSJdLnZhbHVlX2NvdW50cygpLnNvcnRfaW5kZXgoKQogICAgcm93cyA9IFt7ImFyaXR5IjogaW50KGEpLCAiY291bnQiOiBpbnQoYyl9IGZvciBhLCBjIGluIHZjLml0ZW1zKCldCiAgICByZXR1cm4gcGQuRGF0YUZyYW1lKHJvd3MpCgoKIyAtLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLQojIDMuIHBlci1xdWFsaWZpZXIgYWxpZ25tZW50IGZ1bmN0aW9ucyAgKHVzZWQgYnkgdGhlIFFPQSByZXJhbmtlcikKIyAtLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLQpkZWYgbWF0Y2hfanVyaXNkaWN0aW9uKHE6IGRpY3QsIGE6IGRpY3QpIC0+IGZsb2F0OgogICAgIiIiMS4wIHNhbWUgY291bnRyeStyZWdpb24sIDAuNiBzYW1lIGNvdW50cnksIDAgb3RoZXJ3aXNlLiBVbmtub3duIC0+IDAuIiIiCiAgICBxYywgYWMgPSBxLmdldCgiY291bnRyeSIpLCBhLmdldCgiY291bnRyeSIpCiAgICBpZiBub3QgcWMgb3Igbm90IGFjOgogICAgICAgIHJldHVybiAwLjAKICAgIGlmIHFjICE9IGFjOgogICAgICAgIHJldHVybiAwLjAKICAgIHFyLCBhciA9IChxLmdldCgicmVnaW9uIikgb3IgIiIpLCAoYS5nZXQoInJlZ2lvbiIpIG9yICIiKQogICAgcmV0dXJuIDEuMCBpZiAocXIgYW5kIHFyID09IGFyKSBlbHNlIDAuNgoKCmRlZiBtYXRjaF90b3BpYyhxOiBkaWN0LCBhOiBkaWN0KSAtPiBmbG9hdDoKICAgICIiIkV4YWN0IHN1YnRvcGljIDEuMDsgc2FtZSBFU0cgbGV0dGVyIDAuNTsgZWxzZSAwLiBVbmtub3duIC0+IDAuIiIiCiAgICBxcywgYXNfID0gKHEuZ2V0KCJzdWJ0b3BpYyIpIG9yICIiKSwgKGEuZ2V0KCJzdWJ0b3BpYyIpIG9yICIiKQogICAgaWYgcXMgYW5kIHFzID09IGFzXzoKICAgICAgICByZXR1cm4gMS4wCiAgICBxZSwgYWUgPSAocS5nZXQoImVzZyIpIG9yICIiKSwgKGEuZ2V0KCJlc2ciKSBvciAiIikKICAgIGlmIHFlIGFuZCBxZSA9PSBhZToKICAgICAgICByZXR1cm4gMC41CiAgICByZXR1cm4gMC4wCgoKZGVmIG1hdGNoX3RlbXBvcmFsKHE6IGRpY3QsIGE6IGRpY3QsIHRhdTogZmxvYXQgPSBRT0FfVEVNUE9SQUxfVEFVKSAtPiBmbG9hdDoKICAgICIiImV4cCgtfGR5fC90YXUpIHllYXIgcHJveGltaXR5LiBVbmtub3duIG9uIGVpdGhlciBzaWRlIC0+IDAgKG5ldXRyYWwpLiIiIgogICAgcXksIGF5ID0gcS5nZXQoInllYXIiKSwgYS5nZXQoInllYXIiKQogICAgaWYgcXkgaXMgTm9uZSBvciBheSBpcyBOb25lOgogICAgICAgIHJldHVybiAwLjAKICAgIHRyeToKICAgICAgICByZXR1cm4gZmxvYXQobnAuZXhwKC1hYnMoaW50KHF5KSAtIGludChheSkpIC8gZmxvYXQodGF1KSkpCiAgICBleGNlcHQgKFR5cGVFcnJvciwgVmFsdWVFcnJvcik6CiAgICAgICAgcmV0dXJuIDAuMAoKCmRlZiBtYXRjaF9zZXZlcml0eShxOiBkaWN0LCBhOiBkaWN0KSAtPiBmbG9hdDoKICAgICIiIk9yZGluYWwgc2V2ZXJpdHkgY2xvc2VuZXNzIGluIFswLDFdLiBVbmtub3duIC0+IDAuIiIiCiAgICBxcywgYXNfID0gcS5nZXQoInNldmVyaXR5IiksIGEuZ2V0KCJzZXZlcml0eSIpCiAgICBpZiBub3QgcXMgb3Igbm90IGFzXyBvciBxcyBub3QgaW4gU0VWRVJJVFlfT1JERVIgb3IgYXNfIG5vdCBpbiBTRVZFUklUWV9PUkRFUjoKICAgICAgICByZXR1cm4gMC4wCiAgICBkID0gYWJzKFNFVkVSSVRZX09SREVSLmluZGV4KHFzKSAtIFNFVkVSSVRZX09SREVSLmluZGV4KGFzXykpCiAgICByZXR1cm4gMS4wIC0gZCAvIChsZW4oU0VWRVJJVFlfT1JERVIpIC0gMSkKCgpkZWYgbWF0Y2hfZG9jdHlwZShxOiBkaWN0LCBhOiBkaWN0KSAtPiBmbG9hdDoKICAgICIiIjEuMCBpZiBzYW1lIGRvY3VtZW50IHR5cGUsIGVsc2UgMC4gVW5rbm93biAtPiAwLiIiIgogICAgcWQsIGFkID0gKHEuZ2V0KCJkb2N0eXBlIikgb3IgIiIpLCAoYS5nZXQoImRvY3R5cGUiKSBvciAiIikKICAgIHJldHVybiAxLjAgaWYgKHFkIGFuZCBxZCA9PSBhZCkgZWxzZSAwLjAKCgpNQVRDSEVSUyA9IHsKICAgICJqdXJpc2RpY3Rpb24iOiBtYXRjaF9qdXJpc2RpY3Rpb24sCiAgICAidG9waWMiOiAgICAgICAgbWF0Y2hfdG9waWMsCiAgICAidGVtcG9yYWwiOiAgICAgbWF0Y2hfdGVtcG9yYWwsCiAgICAic2V2ZXJpdHkiOiAgICAgbWF0Y2hfc2V2ZXJpdHksCiAgICAiZG9jdHlwZSI6ICAgICAgbWF0Y2hfZG9jdHlwZSwKfQoKCmRlZiBhbGlnbm1lbnRfdmVjdG9yKHFfcHJvZmlsZTogZGljdCwgYV9wcm9maWxlOiBkaWN0LCBkaW1zOiBsaXN0W3N0cl0pIC0+IG5wLm5kYXJyYXk6CiAgICAiIiJQZXItZGltZW5zaW9uIGFsaWdubWVudCBzY29yZXMgZm9yIHRoZSByZXF1ZXN0ZWQgcXVhbGlmaWVyIGRpbXMuIiIiCiAgICByZXR1cm4gbnAuYXJyYXkoW01BVENIRVJTW2RdKHFfcHJvZmlsZSwgYV9wcm9maWxlKSBmb3IgZCBpbiBkaW1zXSwgZHR5cGU9ZmxvYXQpCg==',
    'attributes': 'IiIiCmF0dHJpYnV0ZXMucHkg4oCUIHByZWRpY3QgYSBxdWVyeSdzIGFuY2hvciBhdHRyaWJ1dGVzIEZST00gVEhFIFFVRVJZIFRFWFQuCgpSZXZpZXdlciAyIChNMyk6ICJleHBsYWluIGhvdyBxdWVyeSBhdHRyaWJ1dGVzIHdlcmUgb2J0YWluZWQuIElmIHRoZXkgY29tZSBmcm9tCnRoZSBnb2xkIGRvY3VtZW50LCBhZGQgYSBjb25kaXRpb24gdGhhdCBwcmVkaWN0cyB0aGVtIGZyb20gcXVlcnkgdGV4dC4iCgpXZSB0aGVyZWZvcmUgc3VwcG9ydCB0d28gYXR0cmlidXRlIHNvdXJjZXMgZm9yIHRoZSBRT0EgcmVyYW5rZXI6CgogICogT1JBQ0xFICAgICA6IGF0dHJpYnV0ZXMgcmVhZCBmcm9tIHRoZSBnb2xkIGRvY3VtZW50J3MgYW5jaG9yIHByb2ZpbGUuCiAgICAgICAgICAgICAgICAgVGhpcyBpcyBhbiB1cHBlciBib3VuZCBhbmQgZGVsaWJlcmF0ZWx5IGxlYWtzOyB3ZSByZXBvcnQgaXQgb25seQogICAgICAgICAgICAgICAgIHRvIHF1YW50aWZ5IHRoZSBoZWFkcm9vbS4KICAqIFBSRURJQ1RFRCAgOiBhdHRyaWJ1dGVzIGluZmVycmVkIGZyb20gdGhlIHF1ZXJ5IHN0cmluZyBhbG9uZSwgd2l0aCBOTyBhY2Nlc3MKICAgICAgICAgICAgICAgICB0byB0aGUgZ29sZCBkb2N1bWVudC4gVGhpcyBpcyB0aGUgaG9uZXN0IG9wZXJhdGluZyBjb25kaXRpb24uCgpUaGUgcHJlZGljdG9yIGlzIGludGVudGlvbmFsbHkgbGlnaHQgYW5kIHJlcHJvZHVjaWJsZToKICAtIGp1cmlzZGljdGlvbiAoY291bnRyeSk6IHJ1bGUtYmFzZWQga2V5d29yZCBkZXRlY3Rpb24gKGNvdW50cnkgbmFtZXMgaW4ga28vZW4vdmkvbXMpCiAgLSB0b3BpYyAvIHN1YnRvcGljIC8geWVhcjogMS1OTiBvdmVyIFRSQUlOSU5HIChxdWVyeSAtPiBnb2xkIGFuY2hvcikgcGFpcnMgdXNpbmcKICAgIHRoZSBzYW1lIGNoYXIgbi1ncmFtIFRGLUlERiBhcyB0aGUgcmV0cmlldmVyOyB5ZWFyIGFsc28gdHJpZWQgdmlhIHJlZ2V4LgpObyBleHRlcm5hbCBtb2RlbCBvciBpbnRlcm5ldCByZXF1aXJlZC4KIiIiCmZyb20gX19mdXR1cmVfXyBpbXBvcnQgYW5ub3RhdGlvbnMKCmltcG9ydCByZQppbXBvcnQgbnVtcHkgYXMgbnAKaW1wb3J0IHBhbmRhcyBhcyBwZApmcm9tIHNrbGVhcm4uZmVhdHVyZV9leHRyYWN0aW9uLnRleHQgaW1wb3J0IFRmaWRmVmVjdG9yaXplcgpmcm9tIHNrbGVhcm4ubWV0cmljcy5wYWlyd2lzZSBpbXBvcnQgY29zaW5lX3NpbWlsYXJpdHkKCmZyb20gb250b2xvZ3kgaW1wb3J0IGV4dHJhY3RfeWVhcgoKX0NPVU5UUllfS0VZV09SRFMgPSB7CiAgICAiVmlldG5hbSI6ICBbIuuyoO2KuOuCqCIsICJ2aWV0bmFtIiwgInZpZXQgbmFtIiwgInZp4buHdCIsICJ2aWV0Il0sCiAgICAiTWFsYXlzaWEiOiBbIuunkOugiOydtOyLnOyVhCIsICLrp5DroIjsnbQiLCAibWFsYXlzaWEiLCAibWFsYXkiXSwKfQoKCmRlZiBkZXRlY3RfY291bnRyeSh0ZXh0OiBzdHIpIC0+IHN0cjoKICAgIHQgPSBzdHIodGV4dCBvciAiIikubG93ZXIoKQogICAgZm9yIGNvdW50cnksIGt3cyBpbiBfQ09VTlRSWV9LRVlXT1JEUy5pdGVtcygpOgogICAgICAgIGlmIGFueShrdyBpbiB0IGZvciBrdyBpbiBrd3MpOgogICAgICAgICAgICByZXR1cm4gY291bnRyeQogICAgcmV0dXJuICIiCgoKY2xhc3MgUXVlcnlBdHRyaWJ1dGVQcmVkaWN0b3I6CiAgICAiIiJQcmVkaWN0IHtjb3VudHJ5LCBlc2csIHN1YnRvcGljLCB5ZWFyfSBmb3IgYSBxdWVyeSBmcm9tIGl0cyB0ZXh0IG9ubHkuIiIiCgogICAgZGVmIF9faW5pdF9fKHNlbGYsIG5ncmFtPSgzLCA0KSwgbWF4X2ZlYXR1cmVzPTEwMF8wMDAsIG1pbl9kZj0yKToKICAgICAgICBzZWxmLnZlYyA9IFRmaWRmVmVjdG9yaXplcihhbmFseXplcj0iY2hhcl93YiIsIG5ncmFtX3JhbmdlPW5ncmFtLAogICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgIG1pbl9kZj1taW5fZGYsIG1heF9mZWF0dXJlcz1tYXhfZmVhdHVyZXMsCiAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgZHR5cGU9bnAuZmxvYXQzMikKICAgICAgICBzZWxmLl9maXQgPSBGYWxzZQoKICAgIGRlZiBmaXQoc2VsZiwgdHJhaW5fcXVlcmllczogbGlzdFtzdHJdLCB0cmFpbl9hbmNob3JzOiBsaXN0W2RpY3RdKToKICAgICAgICAiIiJ0cmFpbl9hbmNob3JzOiBnb2xkLWRvYyBhbmNob3IgZGljdHMgYWxpZ25lZCB0byB0cmFpbl9xdWVyaWVzLiIiIgogICAgICAgIHNlbGYudHJhaW5fbWF0ID0gc2VsZi52ZWMuZml0X3RyYW5zZm9ybShbc3RyKHEpIGZvciBxIGluIHRyYWluX3F1ZXJpZXNdKQogICAgICAgIHNlbGYudHJhaW5fYW5jaG9ycyA9IGxpc3QodHJhaW5fYW5jaG9ycykKICAgICAgICBzZWxmLl9maXQgPSBUcnVlCiAgICAgICAgcmV0dXJuIHNlbGYKCiAgICBkZWYgcHJlZGljdF9vbmUoc2VsZiwgcXVlcnk6IHN0cikgLT4gZGljdDoKICAgICAgICBhc3NlcnQgc2VsZi5fZml0LCAiY2FsbCBmaXQoKSBmaXJzdCIKICAgICAgICBxID0gc3RyKHF1ZXJ5IG9yICIiKQogICAgICAgICMgMS1OTiBvdmVyIHRyYWluaW5nIHF1ZXJpZXMKICAgICAgICBxdiA9IHNlbGYudmVjLnRyYW5zZm9ybShbcV0pCiAgICAgICAgc2ltcyA9IGNvc2luZV9zaW1pbGFyaXR5KHF2LCBzZWxmLnRyYWluX21hdClbMF0KICAgICAgICBubiA9IGludChucC5hcmdtYXgoc2ltcykpIGlmIGxlbihzaW1zKSBlbHNlIDAKICAgICAgICBubl9hbmNob3IgPSBzZWxmLnRyYWluX2FuY2hvcnNbbm5dIGlmIHNlbGYudHJhaW5fYW5jaG9ycyBlbHNlIHt9CiAgICAgICAgIyBqdXJpc2RpY3Rpb246IHByZWZlciBleHBsaWNpdCBrZXl3b3JkIGluIHRoZSBxdWVyeSwgZWxzZSBuZWlnaGJvdXIKICAgICAgICBjb3VudHJ5ID0gZGV0ZWN0X2NvdW50cnkocSkgb3Igbm5fYW5jaG9yLmdldCgiY291bnRyeSIsICIiKQogICAgICAgICMgeWVhcjogcmVnZXggb24gdGhlIHF1ZXJ5IGZpcnN0LCBlbHNlIG5laWdoYm91cgogICAgICAgIHllYXIgPSBleHRyYWN0X3llYXIocSkKICAgICAgICBpZiB5ZWFyIGlzIE5vbmU6CiAgICAgICAgICAgIHllYXIgPSBubl9hbmNob3IuZ2V0KCJ5ZWFyIikKICAgICAgICByZXR1cm4gewogICAgICAgICAgICAiY291bnRyeSI6ICBjb3VudHJ5LAogICAgICAgICAgICAicmVnaW9uIjogICAiIiwgICAgICAgICAgICAgICAgICAgICAgIyByZWdpb24gcmFyZWx5IHN0YXRlZCBpbiBhIHF1ZXJ5CiAgICAgICAgICAgICJlc2ciOiAgICAgIG5uX2FuY2hvci5nZXQoImVzZyIsICIiKSwKICAgICAgICAgICAgInN1YnRvcGljIjogbm5fYW5jaG9yLmdldCgic3VidG9waWMiLCAiIiksCiAgICAgICAgICAgICJ5ZWFyIjogICAgIHllYXIsCiAgICAgICAgICAgICJkb2N0eXBlIjogIG5uX2FuY2hvci5nZXQoImRvY3R5cGUiLCAiIiksCiAgICAgICAgICAgICJzZXZlcml0eSI6IG5uX2FuY2hvci5nZXQoInNldmVyaXR5IiwgIiIpLAogICAgICAgIH0KCiAgICBkZWYgcHJlZGljdChzZWxmLCBxdWVyaWVzOiBsaXN0W3N0cl0pIC0+IGxpc3RbZGljdF06CiAgICAgICAgcmV0dXJuIFtzZWxmLnByZWRpY3Rfb25lKHEpIGZvciBxIGluIHF1ZXJpZXNdCgoKZGVmIGFuY2hvcl9yb3dfdG9fZGljdChyb3cpIC0+IGRpY3Q6CiAgICAiIiJDb252ZXJ0IGFuIGFuY2hvci1wcm9maWxlIHJvdyAoU2VyaWVzL2RpY3QpIHRvIHRoZSBtYXRjaGVyIGRpY3QuIiIiCiAgICBnID0gcm93LmdldCBpZiBpc2luc3RhbmNlKHJvdywgZGljdCkgZWxzZSAobGFtYmRhIGssIGQ9Tm9uZTogcm93W2tdIGlmIGsgaW4gcm93IGVsc2UgZCkKICAgIHJldHVybiB7CiAgICAgICAgImNvdW50cnkiOiAgZygiY291bnRyeSIsICIiKSwKICAgICAgICAicmVnaW9uIjogICBnKCJyZWdpb24iLCAiIiksCiAgICAgICAgImVzZyI6ICAgICAgZygiZXNnIiwgIiIpLAogICAgICAgICJzdWJ0b3BpYyI6IGcoInN1YnRvcGljIiwgIiIpLAogICAgICAgICJ5ZWFyIjogICAgIGcoInllYXIiLCBOb25lKSwKICAgICAgICAiZG9jdHlwZSI6ICBnKCJkb2N0eXBlIiwgIiIpLAogICAgICAgICJzZXZlcml0eSI6IGcoInNldmVyaXR5IiwgIiIpLAogICAgfQo=',
    'retrieval': 'IiIiCnJldHJpZXZhbC5weSDigJQgcmV0cmlldmVycyArIGV2YWx1YXRpb24gaGFybmVzcyBmb3IgdGhlIHJldmlzaW9uIGV4cGVyaW1lbnRzLgoKUmV0cmlldmVycwotLS0tLS0tLS0tCiogVGZpZGZSZXRyaWV2ZXIgOiBjaGFyIG4tZ3JhbSBURi1JREYgKHJlcHJvZHVjaWJsZSwgb2ZmbGluZSkg4oCUIHRoZSBCMiBiYXNlbGluZS4KKiBEZW5zZVJldHJpZXZlciA6IG11bHRpbGluZ3VhbCBzZW50ZW5jZSBlbWJlZGRpbmdzIChtRTUgLyBCR0UtTTMpIOKAlCB0aGUKICAgICAgICAgICAgICAgICAgIG11bHRpbGluZ3VhbCBkZW5zZSBiYXNlbGluZSByZXZpZXdlciAyIChNMywgTGl0KSBhc2tlZCBmb3IuCiAgICAgICAgICAgICAgICAgICBOZWVkcyBgc2VudGVuY2UtdHJhbnNmb3JtZXJzYDsgZGVncmFkZXMgZ3JhY2VmdWxseSBpZiBhYnNlbnQuCiogcW9hX3JlcmFuayAgICAgOiBRdWFsaWZpZWQgT250b2xvZ3kgQW5jaG9yaW5nIHJlcmFua2VyIChzcmMvb250b2xvZ3kucHkpLiBJdAogICAgICAgICAgICAgICAgICAgcmVvcmRlcnMgYSBiYXNlIHJldHJpZXZlcidzIHRvcC1OIHBvb2wgYnkgYSB3ZWlnaHRlZCBzdW0gb2YKICAgICAgICAgICAgICAgICAgIHBlci1xdWFsaWZpZXIgYWxpZ25tZW50LiBXZWlnaHRzIGFyZSB0dW5lZCBvbiBhIGhlbGQtb3V0IHNwbGl0LgoKTGV2ZWxzIChhYmxhdGlvbik6IHRyaXBsZSAoPSBtZXRhZGF0YSBiYXNlbGluZSBCMykgLT4gcXVhZCAtPiBwZW50dXBsZS4KCkV2ZXJ5dGhpbmcgcmV0dXJucyBQRVItUVVFUlkgYXJyYXlzIChoaXRAaywgcmVjaXByb2NhbCByYW5rKSBzbyB3ZSBjYW4gYXR0YWNoCmJvb3RzdHJhcCBjb25maWRlbmNlIGludGVydmFscyB0byBldmVyeSBudW1iZXIgKHJldmlld2VyIDIsIFJlc3VsdHMgUTQgLyBNNSkuCiIiIgpmcm9tIF9fZnV0dXJlX18gaW1wb3J0IGFubm90YXRpb25zCgppbXBvcnQgbnVtcHkgYXMgbnAKaW1wb3J0IHBhbmRhcyBhcyBwZApmcm9tIHNrbGVhcm4uZmVhdHVyZV9leHRyYWN0aW9uLnRleHQgaW1wb3J0IFRmaWRmVmVjdG9yaXplcgpmcm9tIHNrbGVhcm4ubWV0cmljcy5wYWlyd2lzZSBpbXBvcnQgY29zaW5lX3NpbWlsYXJpdHkKCmZyb20gY29uZmlnIGltcG9ydCAoRE9DX0NIQVJfTElNSVQsIFJFVFJJRVZBTF9LX0xJU1QsIEJPT1RTVFJBUF9OLCBCT09UU1RSQVBfQ0ksCiAgICAgICAgICAgICAgICAgICAgUU9BX0FMUEhBX0dSSUQsIFNQTElUX0ZSQUNTKQpmcm9tIG9udG9sb2d5IGltcG9ydCBhbGlnbm1lbnRfdmVjdG9yCmZyb20gYXR0cmlidXRlcyBpbXBvcnQgYW5jaG9yX3Jvd190b19kaWN0CgoKIyA9PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PQojIGNvcnB1cyBjb25zdHJ1Y3Rpb24KIyA9PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PQpkZWYgYnVpbGRfY29ycHVzKGNvbnRlbnQ6IHBkLkRhdGFGcmFtZSwgYW5jaG9yczogcGQuRGF0YUZyYW1lLCBsYW5nOiBzdHIpOgogICAgIiIiUmV0dXJuIChkb2NfaWRzLCBkb2NfdGV4dCwgYW5jaG9yX2J5X2RvY2lkKSBmb3IgYSBnaXZlbiBsYW5ndWFnZS4KCiAgICBjb250ZW50IDogRGF0YUZyYW1lIHdpdGggY29sdW1ucyBpZCwgZG9jX2lkLCBjb250ZW50X3trbyxlbixsb2NhbH0KICAgIGFuY2hvcnMgOiBhbmNob3ItcHJvZmlsZSBEYXRhRnJhbWUgKG9udG9sb2d5LmJ1aWxkX2FuY2hvcl9wcm9maWxlcykKICAgIGxhbmcgICAgOiAna28nIHwgJ2VuJyB8ICdsb2NhbCcKICAgICIiIgogICAgY29sID0gZiJjb250ZW50X3tsYW5nfSIKICAgIGRmID0gY29udGVudC5zb3J0X3ZhbHVlcygiZG9jX2lkIikucmVzZXRfaW5kZXgoZHJvcD1UcnVlKQogICAgZG9jX2lkcyA9IGRmWyJkb2NfaWQiXS50b19udW1weSgpCiAgICBkb2NfdGV4dCA9IGRmW2NvbF0uZmlsbG5hKCIiKS5hc3R5cGUoc3RyKS5zdHIuc2xpY2UoMCwgRE9DX0NIQVJfTElNSVQpLnRvbGlzdCgpCiAgICBhbmMgPSBhbmNob3JzLnNldF9pbmRleCgiZG9jX2lkIikKICAgIGFuY2hvcl9ieV9kb2NpZCA9IHt9CiAgICBmb3IgZGlkIGluIGRvY19pZHM6CiAgICAgICAgaWYgZGlkIGluIGFuYy5pbmRleDoKICAgICAgICAgICAgcm93ID0gYW5jLmxvY1tkaWRdCiAgICAgICAgICAgIGlmIGlzaW5zdGFuY2Uocm93LCBwZC5EYXRhRnJhbWUpOiAgICAgICMgZHVwIGRvY19pZCAtPiB0YWtlIGZpcnN0CiAgICAgICAgICAgICAgICByb3cgPSByb3cuaWxvY1swXQogICAgICAgICAgICBhbmNob3JfYnlfZG9jaWRbZGlkXSA9IGFuY2hvcl9yb3dfdG9fZGljdChyb3cpCiAgICAgICAgZWxzZToKICAgICAgICAgICAgYW5jaG9yX2J5X2RvY2lkW2RpZF0gPSBhbmNob3Jfcm93X3RvX2RpY3Qoe30pCiAgICByZXR1cm4gZG9jX2lkcywgZG9jX3RleHQsIGFuY2hvcl9ieV9kb2NpZAoKCiMgPT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT0KIyByZXRyaWV2ZXJzCiMgPT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT0KY2xhc3MgVGZpZGZSZXRyaWV2ZXI6CiAgICAiIiJSZXByb2R1Y2libGUgY2hhciBuLWdyYW0gVEYtSURGIHJldHJpZXZlciAoaWRlbnRpY2FsIGNvbmZpZyB0byBuYjA0KS4iIiIKCiAgICBkZWYgX19pbml0X18oc2VsZiwgbmdyYW09KDMsIDQpLCBtaW5fZGY9MywgbWF4X2ZlYXR1cmVzPTEwMF8wMDApOgogICAgICAgIHNlbGYudmVjID0gVGZpZGZWZWN0b3JpemVyKGFuYWx5emVyPSJjaGFyX3diIiwgbmdyYW1fcmFuZ2U9bmdyYW0sCiAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgbWluX2RmPW1pbl9kZiwgbWF4X2ZlYXR1cmVzPW1heF9mZWF0dXJlcywKICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICBkdHlwZT1ucC5mbG9hdDMyKQoKICAgIGRlZiBmaXQoc2VsZiwgZG9jX3RleHQpOgogICAgICAgIHNlbGYuZG9jX21hdCA9IHNlbGYudmVjLmZpdF90cmFuc2Zvcm0oZG9jX3RleHQpCiAgICAgICAgcmV0dXJuIHNlbGYKCiAgICBkZWYgc2VhcmNoKHNlbGYsIHF1ZXJpZXMsIHRvcG49MTAwLCBiYXRjaD0yNTYpOgogICAgICAgICIiIlJldHVybiAoY2FuZF9pZHhbbl9xLHRvcG5dLCBjYW5kX3Njb3JlW25fcSx0b3BuXSkgYnkgYmFzZSBzaW1pbGFyaXR5LiIiIgogICAgICAgIG5fZG9jID0gc2VsZi5kb2NfbWF0LnNoYXBlWzBdCiAgICAgICAgdG9wbiA9IG1pbih0b3BuLCBuX2RvYykKICAgICAgICBhbGxfaWR4LCBhbGxfc2NvcmUgPSBbXSwgW10KICAgICAgICBmb3IgcyBpbiByYW5nZSgwLCBsZW4ocXVlcmllcyksIGJhdGNoKToKICAgICAgICAgICAgcXYgPSBzZWxmLnZlYy50cmFuc2Zvcm0ocXVlcmllc1tzOnMgKyBiYXRjaF0pCiAgICAgICAgICAgIHNpbXMgPSBjb3NpbmVfc2ltaWxhcml0eShxdiwgc2VsZi5kb2NfbWF0KSAgICAgICAgIyAoYiwgbl9kb2MpCiAgICAgICAgICAgIHBhcnQgPSBucC5hcmdwYXJ0aXRpb24oLXNpbXMsIGt0aD10b3BuIC0gMSwgYXhpcz0xKVs6LCA6dG9wbl0KICAgICAgICAgICAgZm9yIGkgaW4gcmFuZ2Uoc2ltcy5zaGFwZVswXSk6CiAgICAgICAgICAgICAgICBjYW5kID0gcGFydFtpXQogICAgICAgICAgICAgICAgY2FuZCA9IGNhbmRbbnAuYXJnc29ydCgtc2ltc1tpLCBjYW5kXSldCiAgICAgICAgICAgICAgICBhbGxfaWR4LmFwcGVuZChjYW5kKQogICAgICAgICAgICAgICAgYWxsX3Njb3JlLmFwcGVuZChzaW1zW2ksIGNhbmRdKQogICAgICAgIHJldHVybiBucC52c3RhY2soYWxsX2lkeCksIG5wLnZzdGFjayhhbGxfc2NvcmUpCgoKY2xhc3MgRGVuc2VSZXRyaWV2ZXI6CiAgICAiIiJNdWx0aWxpbmd1YWwgZGVuc2UgcmV0cmlldmVyIChtRTUgLyBCR0UtTTMpLiBOZWVkcyBzZW50ZW5jZS10cmFuc2Zvcm1lcnMuIiIiCgogICAgZGVmIF9faW5pdF9fKHNlbGYsIG1vZGVsX25hbWUsIGU1X3ByZWZpeGVzPU5vbmUsIG5vcm1hbGl6ZT1UcnVlLCBkZXZpY2U9Tm9uZSk6CiAgICAgICAgZnJvbSBzZW50ZW5jZV90cmFuc2Zvcm1lcnMgaW1wb3J0IFNlbnRlbmNlVHJhbnNmb3JtZXIgICAjIGxhenkgaW1wb3J0CiAgICAgICAgc2VsZi5tb2RlbCA9IFNlbnRlbmNlVHJhbnNmb3JtZXIobW9kZWxfbmFtZSwgZGV2aWNlPWRldmljZSkKICAgICAgICBzZWxmLm1vZGVsX25hbWUgPSBtb2RlbF9uYW1lCiAgICAgICAgc2VsZi5ub3JtYWxpemUgPSBub3JtYWxpemUKICAgICAgICAjIGU1IGZhbWlseSBleHBlY3RzICJxdWVyeToiLyJwYXNzYWdlOiIgcHJlZml4ZXM7IGJnZS1tMyBkb2VzIG5vdC4KICAgICAgICBpZiBlNV9wcmVmaXhlcyBpcyBOb25lOgogICAgICAgICAgICBlNV9wcmVmaXhlcyA9ICJlNSIgaW4gbW9kZWxfbmFtZS5sb3dlcigpCiAgICAgICAgc2VsZi5lNSA9IGU1X3ByZWZpeGVzCgogICAgZGVmIF9lbmNvZGUoc2VsZiwgdGV4dHMsIGtpbmQsIGJhdGNoX3NpemU9NjQpOgogICAgICAgIGlmIHNlbGYuZTU6CiAgICAgICAgICAgIHByZSA9ICJxdWVyeTogIiBpZiBraW5kID09ICJxdWVyeSIgZWxzZSAicGFzc2FnZTogIgogICAgICAgICAgICB0ZXh0cyA9IFtwcmUgKyB0IGZvciB0IGluIHRleHRzXQogICAgICAgIGVtYiA9IHNlbGYubW9kZWwuZW5jb2RlKHRleHRzLCBiYXRjaF9zaXplPWJhdGNoX3NpemUsCiAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgY29udmVydF90b19udW1weT1UcnVlLCBzaG93X3Byb2dyZXNzX2Jhcj1GYWxzZSwKICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICBub3JtYWxpemVfZW1iZWRkaW5ncz1zZWxmLm5vcm1hbGl6ZSkKICAgICAgICByZXR1cm4gZW1iLmFzdHlwZShucC5mbG9hdDMyKQoKICAgIGRlZiBmaXQoc2VsZiwgZG9jX3RleHQpOgogICAgICAgIHNlbGYuZG9jX2VtYiA9IHNlbGYuX2VuY29kZShkb2NfdGV4dCwga2luZD0icGFzc2FnZSIpCiAgICAgICAgcmV0dXJuIHNlbGYKCiAgICBkZWYgc2VhcmNoKHNlbGYsIHF1ZXJpZXMsIHRvcG49MTAwLCBiYXRjaD0yNTYpOgogICAgICAgIG5fZG9jID0gc2VsZi5kb2NfZW1iLnNoYXBlWzBdCiAgICAgICAgdG9wbiA9IG1pbih0b3BuLCBuX2RvYykKICAgICAgICBhbGxfaWR4LCBhbGxfc2NvcmUgPSBbXSwgW10KICAgICAgICBmb3IgcyBpbiByYW5nZSgwLCBsZW4ocXVlcmllcyksIGJhdGNoKToKICAgICAgICAgICAgcWUgPSBzZWxmLl9lbmNvZGUocXVlcmllc1tzOnMgKyBiYXRjaF0sIGtpbmQ9InF1ZXJ5IikKICAgICAgICAgICAgc2ltcyA9IHFlIEAgc2VsZi5kb2NfZW1iLlQgICAgICAgICAgICAgICAgICAgICAgICAgICMgKGIsIG5fZG9jKQogICAgICAgICAgICBwYXJ0ID0gbnAuYXJncGFydGl0aW9uKC1zaW1zLCBrdGg9dG9wbiAtIDEsIGF4aXM9MSlbOiwgOnRvcG5dCiAgICAgICAgICAgIGZvciBpIGluIHJhbmdlKHNpbXMuc2hhcGVbMF0pOgogICAgICAgICAgICAgICAgY2FuZCA9IHBhcnRbaV0KICAgICAgICAgICAgICAgIGNhbmQgPSBjYW5kW25wLmFyZ3NvcnQoLXNpbXNbaSwgY2FuZF0pXQogICAgICAgICAgICAgICAgYWxsX2lkeC5hcHBlbmQoY2FuZCkKICAgICAgICAgICAgICAgIGFsbF9zY29yZS5hcHBlbmQoc2ltc1tpLCBjYW5kXSkKICAgICAgICByZXR1cm4gbnAudnN0YWNrKGFsbF9pZHgpLCBucC52c3RhY2soYWxsX3Njb3JlKQoKCiMgPT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT0KIyBRT0EgcmVyYW5raW5nCiMgPT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT0KZGVmIF9taW5tYXgoeCk6CiAgICB4ID0gbnAuYXNhcnJheSh4LCBkdHlwZT1mbG9hdCkKICAgIGxvLCBoaSA9IHgubWluKCksIHgubWF4KCkKICAgIHJldHVybiAoeCAtIGxvKSAvIChoaSAtIGxvKSBpZiBoaSA+IGxvIGVsc2UgbnAuemVyb3NfbGlrZSh4KQoKCmRlZiBxb2FfcmVyYW5rKGNhbmRfaWR4LCBjYW5kX3Njb3JlLCBxdWVyeV9wcm9maWxlcywgZG9jX2lkcywgYW5jaG9yX2J5X2RvY2lkLAogICAgICAgICAgICAgICBkaW1zLCBhbHBoYSk6CiAgICAiIiJSZW9yZGVyIGVhY2ggcXVlcnkncyBjYW5kaWRhdGUgcG9vbCBieSBiYXNlICsgc3VtX2sgYWxwaGFfayAqIGFsaWduX2suCgogICAgUGFyYW1ldGVycwogICAgLS0tLS0tLS0tLQogICAgY2FuZF9pZHggICAgIDogKG5fcSwgdG9wbikgYmFzZS1yZXRyaWV2ZXIgY2FuZGlkYXRlIGRvYyBpbmRpY2VzCiAgICBjYW5kX3Njb3JlICAgOiAobl9xLCB0b3BuKSBiYXNlIHNpbWlsYXJpdHkgc2NvcmVzCiAgICBxdWVyeV9wcm9maWxlcyA6IGxpc3Qgb2Ygbl9xIG1hdGNoZXIgZGljdHMgKG9yYWNsZSBvciBwcmVkaWN0ZWQpCiAgICBkb2NfaWRzICAgICAgOiAobl9kb2MsKSBhcnJheSBtYXBwaW5nIGluZGV4IC0+IGRvY19pZAogICAgYW5jaG9yX2J5X2RvY2lkIDogZGljdCBkb2NfaWQgLT4gYW5jaG9yIG1hdGNoZXIgZGljdAogICAgZGltcyAgICAgICAgIDogcXVhbGlmaWVyIGRpbWVuc2lvbnMgdG8gdXNlIChRT0EgbGV2ZWwpCiAgICBhbHBoYSAgICAgICAgOiBhcnJheSBhbGlnbmVkIHRvIGBkaW1zYAoKICAgIFJldHVybnMKICAgIC0tLS0tLS0KICAgIHJlcmFua2VkX2lkeCA6IChuX3EsIHRvcG4pIGNhbmRpZGF0ZSBpbmRpY2VzIGluIG5ldyBvcmRlcgogICAgIiIiCiAgICBhbHBoYSA9IG5wLmFzYXJyYXkoYWxwaGEsIGR0eXBlPWZsb2F0KQogICAgbl9xLCB0b3BuID0gY2FuZF9pZHguc2hhcGUKICAgIHJlcmFua2VkID0gbnAuZW1wdHlfbGlrZShjYW5kX2lkeCkKICAgIGZvciBpIGluIHJhbmdlKG5fcSk6CiAgICAgICAgYmFzZSA9IF9taW5tYXgoY2FuZF9zY29yZVtpXSkgICAgICAgICAgICAgICAgICAgICAgIyBzY2FsZSBiYXNlIHRvIFswLDFdCiAgICAgICAgcXAgPSBxdWVyeV9wcm9maWxlc1tpXQogICAgICAgIGJvbnVzID0gbnAuemVyb3ModG9wbikKICAgICAgICBmb3IgaiBpbiByYW5nZSh0b3BuKToKICAgICAgICAgICAgZGlkID0gZG9jX2lkc1tjYW5kX2lkeFtpLCBqXV0KICAgICAgICAgICAgYXYgPSBhbGlnbm1lbnRfdmVjdG9yKHFwLCBhbmNob3JfYnlfZG9jaWQuZ2V0KGRpZCwge30pLCBkaW1zKQogICAgICAgICAgICBib251c1tqXSA9IGZsb2F0KG5wLmRvdChhbHBoYSwgYXYpKQogICAgICAgIGZpbmFsID0gYmFzZSArIGJvbnVzCiAgICAgICAgb3JkZXIgPSBucC5hcmdzb3J0KC1maW5hbCkKICAgICAgICByZXJhbmtlZFtpXSA9IGNhbmRfaWR4W2ldW29yZGVyXQogICAgcmV0dXJuIHJlcmFua2VkCgoKIyA9PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PQojIGV2YWx1YXRpb24gKHdpdGggcGVyLXF1ZXJ5IGFycmF5cyBmb3IgYm9vdHN0cmFwIENJcykKIyA9PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PQpkZWYgZ29sZF9wb3NpdGlvbnMocmFua2VkX2lkeCwgZG9jX2lkcywgZ29sZF9pZHMpOgogICAgIiIiMC1iYXNlZCByYW5rIG9mIGVhY2ggZ29sZCBkb2Mgd2l0aGluIHRoZSByYW5rZWQgY2FuZGlkYXRlIGxpc3Q7IE5vbmUgaWYgYWJzZW50LiIiIgogICAgcG9zID0gW10KICAgIGZvciBpLCBnaWQgaW4gZW51bWVyYXRlKGdvbGRfaWRzKToKICAgICAgICByb3cgPSBkb2NfaWRzW3JhbmtlZF9pZHhbaV1dCiAgICAgICAgaGl0ID0gbnAud2hlcmUocm93ID09IGdpZClbMF0KICAgICAgICBwb3MuYXBwZW5kKGludChoaXRbMF0pIGlmIGxlbihoaXQpIGVsc2UgTm9uZSkKICAgIHJldHVybiBwb3MKCgpkZWYgbWV0cmljc19mcm9tX3Bvc2l0aW9ucyhwb3NpdGlvbnMsIGtfbGlzdD1SRVRSSUVWQUxfS19MSVNULCBuX2Jvb3Q9Qk9PVFNUUkFQX04sCiAgICAgICAgICAgICAgICAgICAgICAgICAgIGNpPUJPT1RTVFJBUF9DSSwgc2VlZD00Mik6CiAgICAiIiJBZ2dyZWdhdGUgcmV0cmlldmFsIG1ldHJpY3MgKyBib290c3RyYXAgQ0lzIGZyb20gcGVyLXF1ZXJ5IGdvbGQgcG9zaXRpb25zLiIiIgogICAgZnJvbSBtZXRyaWNzIGltcG9ydCBib290c3RyYXBfY2kKICAgIG4gPSBsZW4ocG9zaXRpb25zKQogICAgaGl0ID0ge2s6IG5wLmFycmF5KFsxLjAgaWYgKHAgaXMgbm90IE5vbmUgYW5kIHAgPCBrKSBlbHNlIDAuMCBmb3IgcCBpbiBwb3NpdGlvbnNdKQogICAgICAgICAgIGZvciBrIGluIGtfbGlzdH0KICAgIHJyID0gbnAuYXJyYXkoWzEuMCAvIChwICsgMSkgaWYgcCBpcyBub3QgTm9uZSBlbHNlIDAuMCBmb3IgcCBpbiBwb3NpdGlvbnNdKQogICAgbWF4ayA9IG1heChrX2xpc3QpCiAgICBuZGNnID0gbnAuYXJyYXkoWzEuMCAvIG5wLmxvZzIocCArIDIpIGlmIChwIGlzIG5vdCBOb25lIGFuZCBwIDwgbWF4aykgZWxzZSAwLjAKICAgICAgICAgICAgICAgICAgICAgZm9yIHAgaW4gcG9zaXRpb25zXSkKICAgIG91dCA9IHsibiI6IG59CiAgICBmb3IgayBpbiBrX2xpc3Q6CiAgICAgICAgY2lfayA9IGJvb3RzdHJhcF9jaShoaXRba10sIG5fYm9vdD1uX2Jvb3QsIGNpPWNpLCBzZWVkPXNlZWQpCiAgICAgICAgb3V0W2YiUmVjYWxsQHtrfSJdID0gcm91bmQoY2lfa1sicG9pbnQiXSwgNCkKICAgICAgICBvdXRbZiJSZWNhbGxAe2t9X2xvIl0gPSByb3VuZChjaV9rWyJsbyJdLCA0KQogICAgICAgIG91dFtmIlJlY2FsbEB7a31faGkiXSA9IHJvdW5kKGNpX2tbImhpIl0sIDQpCiAgICBtcnJfY2kgPSBib290c3RyYXBfY2kocnIsIG5fYm9vdD1uX2Jvb3QsIGNpPWNpLCBzZWVkPXNlZWQpCiAgICBvdXRbIk1SUiJdID0gcm91bmQobXJyX2NpWyJwb2ludCJdLCA0KQogICAgb3V0WyJNUlJfbG8iXSA9IHJvdW5kKG1ycl9jaVsibG8iXSwgNCkKICAgIG91dFsiTVJSX2hpIl0gPSByb3VuZChtcnJfY2lbImhpIl0sIDQpCiAgICBvdXRbZiJuRENHQHttYXhrfSJdID0gcm91bmQoZmxvYXQobmRjZy5tZWFuKCkpLCA0KQogICAgcmV0dXJuIG91dCwgeyJoaXQiOiBoaXQsICJyciI6IHJyLCAibmRjZyI6IG5kY2d9CgoKIyA9PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PQojIGRldGVybWluaXN0aWMgc3BsaXRzICsgYWxwaGEgdHVuaW5nCiMgPT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT09PT0KZGVmIG1ha2Vfc3BsaXRzKG4sIGZyYWNzPVNQTElUX0ZSQUNTLCBzZWVkPTQyKToKICAgICIiIkRldGVybWluaXN0aWMgaW5kZXggc3BsaXQgaW50byB0cmFpbi92YWwvdGVzdC4iIiIKICAgIHJuZyA9IG5wLnJhbmRvbS5kZWZhdWx0X3JuZyhzZWVkKQogICAgaWR4ID0gbnAuYXJhbmdlKG4pCiAgICBybmcuc2h1ZmZsZShpZHgpCiAgICBuX3RyID0gaW50KGZyYWNzWyJ0cmFpbiJdICogbikKICAgIG5fdmEgPSBpbnQoZnJhY3NbInZhbCJdICogbikKICAgIHJldHVybiB7InRyYWluIjogbnAuc29ydChpZHhbOm5fdHJdKSwKICAgICAgICAgICAgInZhbCI6ICAgbnAuc29ydChpZHhbbl90cjpuX3RyICsgbl92YV0pLAogICAgICAgICAgICAidGVzdCI6ICBucC5zb3J0KGlkeFtuX3RyICsgbl92YTpdKX0KCgpkZWYgdHVuZV9hbHBoYShjYW5kX2lkeCwgY2FuZF9zY29yZSwgcXVlcnlfcHJvZmlsZXMsIGRvY19pZHMsIGFuY2hvcl9ieV9kb2NpZCwKICAgICAgICAgICAgICAgZ29sZF9pZHMsIGRpbXMsIHZhbF9pZHgsIGdyaWQ9UU9BX0FMUEhBX0dSSUQsIHRhcmdldD0iUmVjYWxsQDEwIiwKICAgICAgICAgICAgICAgc2VlZD00Mik6CiAgICAiIiJDb29yZGluYXRlLWFzY2VudCBvdmVyIHBlci1kaW1lbnNpb24gYWxwaGEgb24gdGhlIFZBTElEQVRJT04gc3BsaXQuCgogICAgUmV0dXJucyB0aGUgYWxwaGEgdmVjdG9yIChhbGlnbmVkIHRvIGBkaW1zYCkgbWF4aW1pc2luZyBgdGFyZ2V0YCBvbiB2YWwuCiAgICAiIiIKICAgIGFscGhhID0gbnAuemVyb3MobGVuKGRpbXMpKQoKICAgIGRlZiBzY29yZShhbHBoYV92ZWMpOgogICAgICAgIHJyID0gcW9hX3JlcmFuayhjYW5kX2lkeFt2YWxfaWR4XSwgY2FuZF9zY29yZVt2YWxfaWR4XSwKICAgICAgICAgICAgICAgICAgICAgICAgW3F1ZXJ5X3Byb2ZpbGVzW2ldIGZvciBpIGluIHZhbF9pZHhdLAogICAgICAgICAgICAgICAgICAgICAgICBkb2NfaWRzLCBhbmNob3JfYnlfZG9jaWQsIGRpbXMsIGFscGhhX3ZlYykKICAgICAgICBwb3MgPSBnb2xkX3Bvc2l0aW9ucyhyciwgZG9jX2lkcywgW2dvbGRfaWRzW2ldIGZvciBpIGluIHZhbF9pZHhdKQogICAgICAgIG0sIF8gPSBtZXRyaWNzX2Zyb21fcG9zaXRpb25zKHBvcywgbl9ib290PTEsIHNlZWQ9c2VlZCkgICMgcG9pbnQgb25seQogICAgICAgIHJldHVybiBtW3RhcmdldF0KCiAgICBiZXN0ID0gc2NvcmUoYWxwaGEpCiAgICBpbXByb3ZlZCA9IFRydWUKICAgIHdoaWxlIGltcHJvdmVkOgogICAgICAgIGltcHJvdmVkID0gRmFsc2UKICAgICAgICBmb3IgZCBpbiByYW5nZShsZW4oZGltcykpOgogICAgICAgICAgICBjdXIgPSBhbHBoYVtkXQogICAgICAgICAgICBmb3IgZyBpbiBncmlkOgogICAgICAgICAgICAgICAgaWYgZyA9PSBjdXI6CiAgICAgICAgICAgICAgICAgICAgY29udGludWUKICAgICAgICAgICAgICAgIHRyaWFsID0gYWxwaGEuY29weSgpCiAgICAgICAgICAgICAgICB0cmlhbFtkXSA9IGcKICAgICAgICAgICAgICAgIHMgPSBzY29yZSh0cmlhbCkKICAgICAgICAgICAgICAgIGlmIHMgPiBiZXN0ICsgMWUtOToKICAgICAgICAgICAgICAgICAgICBiZXN0LCBhbHBoYSwgaW1wcm92ZWQgPSBzLCB0cmlhbCwgVHJ1ZQogICAgcmV0dXJuIGFscGhhLCBiZXN0Cg==',
    'llm': 'IiIiCmxsbS5weSDigJQgc2hhcmVkIE9wZW5BSSBqdWRnZTogY2FjaGVkIGxsbV9jYWxsKCkgKyBDb3N0VHJhY2tlciArIG1vZGVsIHRpZXJzLgoKRXh0cmFjdGVkIGZyb20gbm90ZWJvb2sgMDUgQ2VsbCBBIHNvIHRoYXQgQk9USCB0aGUgY3Jvc3MtbGluZ3VhbCBqdWRnbWVudAoobmIwNSkgYW5kIHRoZSBldmlkZW5jZS1jb25kaXRpb25lZCBqdWRnbWVudCAvIGJvdHRsZW5lY2sgZXhwZXJpbWVudCAobmIwOSkKdXNlIHRoZSAqaWRlbnRpY2FsKiBjYWxsIHBhdGggYW5kIGNhY2hlLiBUaGlzIG1hdHRlcnMgZm9yIHRoZSByZXZpc2lvbjoKcmV2aWV3ZXIgMiAoTTIpIGFza3MgdXMgdG8gY29tcGFyZSBqdWRnbWVudHMgb24gdGhlIHNhbWUgY2FzZXMgdW5kZXIKZGlmZmVyZW50IGV2aWRlbmNlOyB1c2luZyBvbmUganVkZ2UgaW1wbGVtZW50YXRpb24gZ3VhcmFudGVlcyB0aGUgb25seSB0aGluZwp0aGF0IGNoYW5nZXMgYWNyb3NzIGNvbmRpdGlvbnMgaXMgdGhlIGV2aWRlbmNlLCBub3QgdGhlIHBsdW1iaW5nLgoKQ2FjaGluZwotLS0tLS0tClJlc3BvbnNlcyBhcmUgY2FjaGVkIHVuZGVyIGFydGlmYWN0cy9sbG1fY2FjaGUvPGtleT4uanNvbiB3aGVyZSBrZXkgaGFzaGVzCihtb2RlbCwgc3lzdGVtLCB1c2VyLCB0ZW1wZXJhdHVyZSwgcmVzcG9uc2VfanNvbikuIFJlLXJ1bm5pbmcgaXMgdGhlcmVmb3JlCmZyZWUgZm9yIGFscmVhZHktc2VlbiBwcm9tcHRzLiBUaGUgc2hpcHBlZCBjYWNoZSBjb3ZlcnMgdGhlIFdFQUsgdGllciBvbmx5OwptaWQvc3Ryb25nIHRpZXJzIGFuZCBhbnkgbmV3IChldmlkZW5jZS1jb25kaXRpb25lZCkgcHJvbXB0IHdpbGwgaXNzdWUgbGl2ZQpjYWxscyBvbiB0aGUgdXNlcidzIG1hY2hpbmUgd2l0aCB0aGVpciBvd24gT1BFTkFJX0FQSV9LRVkuCiIiIgpmcm9tIF9fZnV0dXJlX18gaW1wb3J0IGFubm90YXRpb25zCgppbXBvcnQgaGFzaGxpYgppbXBvcnQganNvbgppbXBvcnQgb3MKaW1wb3J0IHJlCmltcG9ydCB0aW1lCmZyb20gcGF0aGxpYiBpbXBvcnQgUGF0aApmcm9tIHR5cGluZyBpbXBvcnQgQW55Cgp0cnk6CiAgICBmcm9tIGRvdGVudiBpbXBvcnQgbG9hZF9kb3RlbnYKZXhjZXB0IEltcG9ydEVycm9yOiAgIyBkb3RlbnYgaXMgb3B0aW9uYWw7IGVudiB2YXJzIG1heSBiZSBzZXQgYW5vdGhlciB3YXkKICAgIGRlZiBsb2FkX2RvdGVudigqX2EsICoqX2spOiAgIyB0eXBlOiBpZ25vcmUKICAgICAgICByZXR1cm4gRmFsc2UKCmZyb20gY29uZmlnIGltcG9ydCBQUk9KRUNUX1JPT1QKCkNBQ0hFX0RJUiA9IFBST0pFQ1RfUk9PVCAvICJhcnRpZmFjdHMiIC8gImxsbV9jYWNoZSIKQ0FDSEVfRElSLm1rZGlyKHBhcmVudHM9VHJ1ZSwgZXhpc3Rfb2s9VHJ1ZSkKCgojIC0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tCiMgbW9kZWwgdGllcnMgKG5hbWVzIGNvbWUgZnJvbSAuZW52OyBwcmljaW5nIGZyb20gdGhlIHVzZXIncyBtYW5pZmVzdCkKIyAtLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLQpkZWYgbG9hZF9tb2RlbHMoKSAtPiBkaWN0OgogICAgbG9hZF9kb3RlbnYoUFJPSkVDVF9ST09UIC8gIi5lbnYiKQogICAgbG9hZF9kb3RlbnYoUGF0aC5jd2QoKSAvICIuZW52IikKICAgIHJldHVybiB7CiAgICAgICAgIndlYWsiOiAgIHsibmFtZSI6IG9zLmdldGVudigiTExNX01PREVMX1dFQUsiLCAiZ3B0LTRvLW1pbmkiKSwKICAgICAgICAgICAgICAgICAgICJpbiI6IDAuMTUsICJjYWNoZWRfaW4iOiAwLjA3NSwgIm91dCI6IDAuNjB9LAogICAgICAgICJtaWQiOiAgICB7Im5hbWUiOiBvcy5nZXRlbnYoIkxMTV9NT0RFTF9NSUQiLCAiZ3B0LTUuNC1taW5pIiksCiAgICAgICAgICAgICAgICAgICAiaW4iOiAwLjc1LCAiY2FjaGVkX2luIjogMC4wNzUsICJvdXQiOiA0LjUwfSwKICAgICAgICAic3Ryb25nIjogeyJuYW1lIjogb3MuZ2V0ZW52KCJMTE1fTU9ERUxfU1RST05HIiwgImdwdC01LjQiKSwKICAgICAgICAgICAgICAgICAgICJpbiI6IE5vbmUsICJjYWNoZWRfaW4iOiBOb25lLCAib3V0IjogTm9uZX0sCiAgICB9CgoKTU9ERUxTID0gbG9hZF9tb2RlbHMoKQoKCmNsYXNzIENvc3RUcmFja2VyOgogICAgZGVmIF9faW5pdF9fKHNlbGYpOgogICAgICAgIHNlbGYucmVjb3JkczogbGlzdFtkaWN0XSA9IFtdCgogICAgZGVmIGFkZChzZWxmLCB0aWVyLCBtb2RlbCwgdXNhZ2UsIHRhZz0iIik6CiAgICAgICAgcCA9IE1PREVMUy5nZXQodGllciwge30pCiAgICAgICAgcGluLCBwY2FjaGVkLCBwb3V0ID0gcC5nZXQoImluIiksIHAuZ2V0KCJjYWNoZWRfaW4iKSwgcC5nZXQoIm91dCIpCiAgICAgICAgcHQgPSBnZXRhdHRyKHVzYWdlLCAicHJvbXB0X3Rva2VucyIsIDApIG9yIDAKICAgICAgICBjdCA9IGdldGF0dHIodXNhZ2UsICJjb21wbGV0aW9uX3Rva2VucyIsIDApIG9yIDAKICAgICAgICBjYWNoZWQgPSAwCiAgICAgICAgZGV0ID0gZ2V0YXR0cih1c2FnZSwgInByb21wdF90b2tlbnNfZGV0YWlscyIsIE5vbmUpCiAgICAgICAgaWYgZGV0IGlzIG5vdCBOb25lOgogICAgICAgICAgICBjYWNoZWQgPSBnZXRhdHRyKGRldCwgImNhY2hlZF90b2tlbnMiLCAwKSBvciAwCiAgICAgICAgZnJlc2ggPSBtYXgocHQgLSBjYWNoZWQsIDApCiAgICAgICAgY29zdCA9IE5vbmUKICAgICAgICBpZiBOb25lIG5vdCBpbiAocGluLCBwb3V0KToKICAgICAgICAgICAgcGMgPSBwY2FjaGVkIGlmIHBjYWNoZWQgaXMgbm90IE5vbmUgZWxzZSBwaW4KICAgICAgICAgICAgY29zdCA9IChmcmVzaCAqIHBpbiArIGNhY2hlZCAqIHBjICsgY3QgKiBwb3V0KSAvIDFfMDAwXzAwMAogICAgICAgIHNlbGYucmVjb3Jkcy5hcHBlbmQoeyJ0YWciOiB0YWcsICJ0aWVyIjogdGllciwgIm1vZGVsIjogbW9kZWwsCiAgICAgICAgICAgICAgICAgICAgICAgICAgICAgInByb21wdF90b2tlbnMiOiBwdCwgImNhY2hlZF90b2tlbnMiOiBjYWNoZWQsCiAgICAgICAgICAgICAgICAgICAgICAgICAgICAgImNvbXBsZXRpb25fdG9rZW5zIjogY3QsICJjb3N0X3VzZCI6IGNvc3R9KQogICAgICAgIHJldHVybiBjb3N0IG9yIDAuMAoKICAgIGRlZiB0b3RhbF91c2Qoc2VsZik6CiAgICAgICAgcmV0dXJuIGZsb2F0KHN1bShyWyJjb3N0X3VzZCJdIG9yIDAuMCBmb3IgciBpbiBzZWxmLnJlY29yZHMpKQoKCkNPU1QgPSBDb3N0VHJhY2tlcigpCgoKZGVmIF9zYWZlX2pzb24odGV4dCk6CiAgICBpZiBub3QgdGV4dDoKICAgICAgICByZXR1cm4gTm9uZQogICAgdCA9IHJlLnN1YihyIl5gYGAoPzpqc29uKT9ccyp8XHMqYGBgJCIsICIiLCB0ZXh0LnN0cmlwKCksIGZsYWdzPXJlLlMpLnN0cmlwKCkKICAgIHRyeToKICAgICAgICByZXR1cm4ganNvbi5sb2Fkcyh0KQogICAgZXhjZXB0IEV4Y2VwdGlvbjoKICAgICAgICBwYXNzCiAgICBmb3Igb3BlbmVyLCBjbG9zZXIgaW4gKCgieyIsICJ9IiksICgiWyIsICJdIikpOgogICAgICAgIGksIGogPSB0LmZpbmQob3BlbmVyKSwgdC5yZmluZChjbG9zZXIpCiAgICAgICAgaWYgMCA8PSBpIDwgajoKICAgICAgICAgICAgdHJ5OgogICAgICAgICAgICAgICAgcmV0dXJuIGpzb24ubG9hZHModFtpOmogKyAxXSkKICAgICAgICAgICAgZXhjZXB0IEV4Y2VwdGlvbjoKICAgICAgICAgICAgICAgIGNvbnRpbnVlCiAgICByZXR1cm4gTm9uZQoKCmRlZiBfY2FjaGVfa2V5KG1vZGVsLCBzeXN0ZW0sIHVzZXIsIHRlbXBlcmF0dXJlLCByZXNwb25zZV9qc29uKToKICAgIHJhdyA9IGpzb24uZHVtcHMoeyJtIjogbW9kZWwsICJzIjogc3lzdGVtLCAidSI6IHVzZXIsCiAgICAgICAgICAgICAgICAgICAgICAidCI6IHRlbXBlcmF0dXJlLCAiaiI6IHJlc3BvbnNlX2pzb259LAogICAgICAgICAgICAgICAgICAgICBlbnN1cmVfYXNjaWk9RmFsc2UsIHNvcnRfa2V5cz1UcnVlKQogICAgcmV0dXJuIGhhc2hsaWIuc2hhMjU2KHJhdy5lbmNvZGUoInV0Zi04IikpLmhleGRpZ2VzdCgpWzoyNF0KCgpkZWYgX2NsaWVudCgpOgogICAgIiIiTGF6eSBPcGVuQUkgY2xpZW50IHNvIHRoYXQgb2ZmbGluZSwgY2FjaGUtb25seSBydW5zIG5lZWQgbm8gQVBJIGtleS4iIiIKICAgIGZyb20gb3BlbmFpIGltcG9ydCBPcGVuQUkKICAgIGtleSA9IG9zLmdldGVudigiT1BFTkFJX0FQSV9LRVkiKQogICAgaWYgbm90IGtleToKICAgICAgICByYWlzZSBWYWx1ZUVycm9yKCJPUEVOQUlfQVBJX0tFWSBub3Qgc2V0IChuZWVkZWQgb25seSBmb3IgY2FjaGUgbWlzc2VzKS4iKQogICAgcmV0dXJuIE9wZW5BSShhcGlfa2V5PWtleSkKCgpkZWYgbGxtX2NhbGwoc3lzdGVtLCB1c2VyLCB0aWVyPSJ3ZWFrIiwgdGVtcGVyYXR1cmU9MC4wLAogICAgICAgICAgICAgcmVzcG9uc2VfanNvbj1UcnVlLCB0YWc9IiIsIHVzZV9jYWNoZT1UcnVlLCBtYXhfcmV0cmllcz0zKToKICAgICIiIkNhY2hlZCBjaGF0IGNhbGwuIFJldHVybnMge3RleHQsIGpzb24sIGNhY2hlZCwgdGllciwgbW9kZWwsIGNvc3RfdXNkfS4KCiAgICBJZGVudGljYWwga2V5L2JlaGF2aW91ciB0byB0aGUgc3VibWl0dGVkIG5iMDUgc28gdGhlIHNoaXBwZWQgY2FjaGUga2VlcHMKICAgIGhpdHRpbmcuIEEgY2FjaGUgTUlTUyB0cmlnZ2VycyBhIGxpdmUgY2FsbCAobmVlZHMgT1BFTkFJX0FQSV9LRVkpLgogICAgIiIiCiAgICBtb2RlbCA9IE1PREVMU1t0aWVyXVsibmFtZSJdCiAgICBrZXkgPSBfY2FjaGVfa2V5KG1vZGVsLCBzeXN0ZW0sIHVzZXIsIHRlbXBlcmF0dXJlLCByZXNwb25zZV9qc29uKQogICAgY2FjaGVfZmlsZSA9IENBQ0hFX0RJUiAvIGYie2tleX0uanNvbiIKICAgIGlmIHVzZV9jYWNoZSBhbmQgY2FjaGVfZmlsZS5leGlzdHMoKToKICAgICAgICBjID0ganNvbi5sb2FkcyhjYWNoZV9maWxlLnJlYWRfdGV4dChlbmNvZGluZz0idXRmLTgiKSkKICAgICAgICBjWyJjYWNoZWQiXSA9IFRydWUKICAgICAgICBvcmlnaW5hbF9jb3N0ID0gYy5nZXQoImNvc3RfdXNkIiwgMC4wKSBvciAwLjAKICAgICAgICBDT1NULnJlY29yZHMuYXBwZW5kKHsidGFnIjogdGFnIG9yIGMuZ2V0KCJtb2RlbCIsICIiKSwKICAgICAgICAgICAgICAgICAgICAgICAgICAgICAidGllciI6IGMuZ2V0KCJ0aWVyIiwgdGllciksICJtb2RlbCI6IGMuZ2V0KCJtb2RlbCIsICIiKSwKICAgICAgICAgICAgICAgICAgICAgICAgICAgICAicHJvbXB0X3Rva2VucyI6IDAsICJjYWNoZWRfdG9rZW5zIjogMCwKICAgICAgICAgICAgICAgICAgICAgICAgICAgICAiY29tcGxldGlvbl90b2tlbnMiOiAwLCAiY29zdF91c2QiOiBvcmlnaW5hbF9jb3N0fSkKICAgICAgICBjWyJjb3N0X3VzZCJdID0gb3JpZ2luYWxfY29zdAogICAgICAgIHJldHVybiBjCgogICAga3dhcmdzOiBkaWN0W3N0ciwgQW55XSA9IHsKICAgICAgICAibW9kZWwiOiBtb2RlbCwKICAgICAgICAibWVzc2FnZXMiOiBbeyJyb2xlIjogInN5c3RlbSIsICJjb250ZW50Ijogc3lzdGVtfSwKICAgICAgICAgICAgICAgICAgICAgeyJyb2xlIjogInVzZXIiLCAiY29udGVudCI6IHVzZXJ9XSwKICAgIH0KICAgIGlmIHJlc3BvbnNlX2pzb246CiAgICAgICAga3dhcmdzWyJyZXNwb25zZV9mb3JtYXQiXSA9IHsidHlwZSI6ICJqc29uX29iamVjdCJ9CiAgICBrd2FyZ3NbInRlbXBlcmF0dXJlIl0gPSB0ZW1wZXJhdHVyZQoKICAgIGNsaWVudCA9IF9jbGllbnQoKQogICAgbGFzdF9lcnIgPSBOb25lCiAgICBmb3IgYXR0ZW1wdCBpbiByYW5nZSgxLCBtYXhfcmV0cmllcyArIDEpOgogICAgICAgIHRyeToKICAgICAgICAgICAgcmVzcCA9IGNsaWVudC5jaGF0LmNvbXBsZXRpb25zLmNyZWF0ZSgqKmt3YXJncykKICAgICAgICAgICAgdGV4dCA9IHJlc3AuY2hvaWNlc1swXS5tZXNzYWdlLmNvbnRlbnQgb3IgIiIKICAgICAgICAgICAgcGFyc2VkID0gX3NhZmVfanNvbih0ZXh0KSBpZiByZXNwb25zZV9qc29uIGVsc2UgTm9uZQogICAgICAgICAgICBpZiByZXNwb25zZV9qc29uIGFuZCBwYXJzZWQgaXMgTm9uZToKICAgICAgICAgICAgICAgIHJhaXNlIFZhbHVlRXJyb3IoIlJlc3BvbnNlIHdhcyBub3QgdmFsaWQgSlNPTi4iKQogICAgICAgICAgICBjb3N0ID0gQ09TVC5hZGQodGllciwgbW9kZWwsIHJlc3AudXNhZ2UsIHRhZz10YWcgb3IgbW9kZWwpCiAgICAgICAgICAgIG91dCA9IHsidGV4dCI6IHRleHQsICJqc29uIjogcGFyc2VkLCAiY2FjaGVkIjogRmFsc2UsCiAgICAgICAgICAgICAgICAgICAidGllciI6IHRpZXIsICJtb2RlbCI6IG1vZGVsLCAiY29zdF91c2QiOiBjb3N0fQogICAgICAgICAgICBpZiB1c2VfY2FjaGU6CiAgICAgICAgICAgICAgICBjYWNoZV9maWxlLndyaXRlX3RleHQoanNvbi5kdW1wcyhvdXQsIGVuc3VyZV9hc2NpaT1GYWxzZSksCiAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgZW5jb2Rpbmc9InV0Zi04IikKICAgICAgICAgICAgcmV0dXJuIG91dAogICAgICAgIGV4Y2VwdCBUeXBlRXJyb3IgYXMgZToKICAgICAgICAgICAgaWYgInRlbXBlcmF0dXJlIiBpbiBrd2FyZ3M6CiAgICAgICAgICAgICAgICBrd2FyZ3MucG9wKCJ0ZW1wZXJhdHVyZSIsIE5vbmUpCiAgICAgICAgICAgICAgICBsYXN0X2VyciA9IGUKICAgICAgICAgICAgICAgIGNvbnRpbnVlCiAgICAgICAgICAgIGxhc3RfZXJyID0gZQogICAgICAgIGV4Y2VwdCBFeGNlcHRpb24gYXMgZToKICAgICAgICAgICAgbGFzdF9lcnIgPSBlCiAgICAgICAgICAgIHRpbWUuc2xlZXAobWluKDIgKiogYXR0ZW1wdCwgOCkpCiAgICByYWlzZSBSdW50aW1lRXJyb3IoZiJbbGxtX2NhbGxdIGZhaWxlZCBhZnRlciB7bWF4X3JldHJpZXN9IHJldHJpZXM6IHtsYXN0X2Vycn0iKQoKCmRlZiB0aWVyX2F2YWlsYWJsZSh0aWVyKSAtPiBib29sOgogICAgIiIiQ2hlYXAgcHJvYmU6IHRyeSBhIDEtdG9rZW4gY2FjaGVkL3JlYWwgY2FsbDsgRmFsc2UgaWYgdW5yZWFjaGFibGUuIiIiCiAgICB0cnk6CiAgICAgICAgbGxtX2NhbGwoInBpbmciLCAnUmV0dXJuIEpTT04geyJvayI6dHJ1ZX0uJywgdGllcj10aWVyLCB0YWc9ZiJwcm9iZS17dGllcn0iKQogICAgICAgIHJldHVybiBUcnVlCiAgICBleGNlcHQgRXhjZXB0aW9uIGFzIGU6CiAgICAgICAgcHJpbnQoZiJbc2tpcF0gdGllciAne3RpZXJ9JyAoe01PREVMU1t0aWVyXVsnbmFtZSddfSkgdW5hdmFpbGFibGU6IHtlfSIpCiAgICAgICAgcmV0dXJuIEZhbHNlCg==',
}
_ORDER = ['config', 'data_loader', 'labeling', 'metrics', 'ontology', 'attributes', 'retrieval', 'llm']
for _n in _ORDER:
    _m = types.ModuleType(_n)
    _m.__dict__["__file__"] = str(_ROOT / "src" / (_n + ".py"))
    sys.modules[_n] = _m
    exec(compile(base64.b64decode(_SRC[_n]).decode("utf-8"), _n + ".py", "exec"), _m.__dict__)
from config import *   # paths, constants, normalize_cat_country, set_seed
# --------------------------------------------------------------------------

# ---- shared setup (paths, seed, plotting) ---------------------------------
import sys, json, warnings
from pathlib import Path
warnings.filterwarnings("ignore")
import numpy as np, pandas as pd
import matplotlib, matplotlib.pyplot as plt
import seaborn as sns

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT / "src"))

from config import (INTERIM_DIR, PROCESSED_DIR, TABLES_DIR, FIGURES_DIR,
                    RETRIEVAL_K_LIST, QOA_LEVELS, QOA_DIMENSIONS, DENSE_MODELS,
                    BOOTSTRAP_N, TOST_MARGIN, SELECTIVE_RISK_LEVELS, set_seed)
set_seed()
matplotlib.rcParams.update({"axes.unicode_minus": False, "figure.dpi": 110})
# Figure convention: seaborn, grayscale, no titles, dpi=600, save PNG + PDF.
sns.set_theme(style="whitegrid", context="paper")
sns.set_palette("Greys_r")
GREY_DARK, GREY_MID, GREY_LIGHT = "#333333", "#777777", "#BBBBBB"
FIG_DPI = 600
def save_fig(fig, stem):
    for ext in ("png", "pdf"):
        fig.savefig(FIGURES_DIR / f"{stem}.{ext}", dpi=FIG_DPI, bbox_inches="tight")
    print("[save] figure", stem)
def save_tbl(df, stem):
    df.to_csv(TABLES_DIR / f"{stem}.csv", index=False, encoding="utf-8-sig")
    print("[save] table", stem, df.shape)


# 1. Judge, retrievers, eval cases
import retrieval, attributes
from data_loader import load_content
from llm import llm_call, tier_available, COST, MODELS

TIERS = ["weak"]           # extend to ["weak","mid","strong"] on your machine
TOPK_EVIDENCE = 3          # docs shown to the judge in retrieved conditions
CHAR_PER_DOC = 2000
N_PER_GROUP = 100          # stratified cases per (country,label); raise for the final run

source = pd.read_parquet(INTERIM_DIR / "source.parquet")
source = normalize_cat_country(source)
gold = pd.read_parquet(PROCESSED_DIR / "compliance_gold.parquet")
gold = normalize_cat_country(gold)
anchors = pd.read_parquet(PROCESSED_DIR / "anchors.parquet")

# cases with their own text (gold-evidence condition)
case_content = load_content(ids=set(gold["id"]))
cc = case_content.set_index("id")
evalset = gold.merge(case_content[["id"]], on="id", how="inner")
def strat(df, by, n, seed=42):
    return pd.concat([g.sample(n=min(n, len(g)), random_state=seed)
                      for _, g in df.groupby(by)]).sort_values("id").reset_index(drop=True)
eval_cases = strat(evalset, ["country", "label"], N_PER_GROUP)
print("eval cases:", len(eval_cases))


# 2. Build a retrieval corpus (regulations + cases) and both retrievers
corpus_meta = source[source.category.isin(CORPUS_CATEGORIES_EN)][["id", "doc_id", "country", "category"]]
corpus_content = load_content(ids=set(corpus_meta["id"]))
corpus_content["doc_id"] = corpus_content["id"].map(dict(zip(corpus_meta["id"], corpus_meta["doc_id"])))
doc_ids, doc_text, anc_by = retrieval.build_corpus(corpus_content, anchors, "ko")
docid_to_text = dict(zip(corpus_content["doc_id"], corpus_content["content_ko"].fillna("")))

# query per case = its detail/title; retrieve top-k evidence
case_meta = source.set_index("id")
def case_query(cid):
    row = case_meta.loc[cid] if cid in case_meta.index else None
    if row is None: return ""
    q = str(row.get("case_detail", "") or "") or str(row.get("title", "") or "")
    return q[:1000]
queries = [case_query(c) for c in eval_cases["id"]]

base = retrieval.TfidfRetriever().fit(doc_text)
cand_idx, cand_score = base.search(queries, topn=50)

# QOA pentuple with predicted attributes (honest condition)
oracle_profiles = []
anc_idx = anchors.set_index("id")
for cid in eval_cases["id"]:
    oracle_profiles.append(attributes.anchor_row_to_dict(anc_idx.loc[cid]) if cid in anc_idx.index
                           else attributes.anchor_row_to_dict({}))
predictor = attributes.QueryAttributePredictor().fit(queries, oracle_profiles)
pred_profiles = predictor.predict(queries)
dims = QOA_LEVELS["pentuple"]
# light alpha: unit weights (final run should load tuned alpha from nb08)
alpha = np.ones(len(dims))
rr = retrieval.qoa_rerank(cand_idx, cand_score, pred_profiles, doc_ids, anc_by, dims, alpha)


# 3. Judge prompts (one judge, three evidence conditions)
SYS = ("You are a legal-compliance analyst. Decide whether legal liability was "
       "established against the defendant. Respond ONLY as JSON: "
       '{"label": one of ["NON_COMPLIANT","COMPLIANT","OTHER"], '
       '"confidence": a number in [0,1]}. NON_COMPLIANT = liability/sanction '
       "established; COMPLIANT = no liability; OTHER = mixed/unclear. Judge only "
       "from the given text.")
VALID = {"NON_COMPLIANT", "COMPLIANT", "OTHER"}
from labeling import mask_verdict   # R1: verdict-masking for the leakage control (M4)

def evidence_text(cond, i, cid):
    if cond in ("gold", "gold_masked"):
        txt = str(cc.loc[cid]["content_ko"])[:TOPK_EVIDENCE * CHAR_PER_DOC] if cid in cc.index else ""
        # gold_masked redacts sanction/acquittal tokens so the judge cannot copy
        # the outcome -> tests whether the task is just verdict extraction (M4).
        return mask_verdict(txt) if cond == "gold_masked" else txt
    order = cand_idx[i] if cond == "retrieved_B2" else rr[i]
    docs = [docid_to_text.get(doc_ids[j], "")[:CHAR_PER_DOC] for j in order[:TOPK_EVIDENCE]]
    return "\n\n---\n\n".join(docs)

def judge(cond, i, cid, tier):
    txt = evidence_text(cond, i, cid)
    user = f"EVIDENCE ({cond}):\n{txt}\n\nReturn the JSON now."
    res = llm_call(SYS, user, tier=tier, temperature=0.0, response_json=True,
                   tag=f"judge-{cond}-{tier}")
    j = res.get("json") or {}
    lab = str(j.get("label", "")).upper().strip()
    if lab not in VALID: lab = "OTHER"
    try: conf = min(1.0, max(0.0, float(j.get("confidence", 0.5))))
    except Exception: conf = 0.5
    return lab, conf


# 4. Run all conditions x tiers, collect predictions, compute accuracy
from metrics import bootstrap_ci
# "gold_masked" = gold text with verdict tokens redacted (M4 leakage control).
CONDS = ["gold", "gold_masked", "retrieved_B2", "retrieved_QOA"]
rows = []; preds_long = []
for tier in TIERS:
    if not tier_available(tier):
        continue
    for cond in CONDS:
        pr = []
        for i, cid in enumerate(eval_cases["id"]):
            lab, conf = judge(cond, i, cid, tier)
            pr.append((lab, conf))
        pred = [p[0] for p in pr]; conf = [p[1] for p in pr]
        gold_lab = eval_cases["label"].tolist()
        correct = np.array([1.0 if p == g else 0.0 for p, g in zip(pred, gold_lab)])
        ci = bootstrap_ci(correct, n_boot=BOOTSTRAP_N)
        # OTHER-separated accuracy: treat model-OTHER as abstention, not a wrong
        # decision. decided_acc = accuracy among non-OTHER predictions; coverage
        # = fraction decided. (gold is NON/COMP only, so raw acc penalises OTHER.)
        decided = np.array([p != "OTHER" for p in pred])
        cov = float(decided.mean())
        dec_acc = float(np.mean([1.0 if pred[k] == gold_lab[k] else 0.0
                                 for k in range(len(pred)) if decided[k]])) if decided.any() else 0.0
        rows.append({"tier": MODELS[tier]["name"], "condition": cond,
                     "accuracy": round(ci["point"], 4), "acc_lo": round(ci["lo"], 4),
                     "acc_hi": round(ci["hi"], 4),
                     "decided_acc": round(dec_acc, 4), "coverage": round(cov, 4),
                     "n": len(correct)})
        for i, cid in enumerate(eval_cases["id"]):
            preds_long.append({"tier": tier, "condition": cond, "id": cid,
                               "country": eval_cases.iloc[i]["country"],
                               "gold": gold_lab[i], "pred": pred[i], "conf": conf[i]})
        print(f"[{tier}/{cond}] acc={ci['point']:.4f} [{ci['lo']:.3f},{ci['hi']:.3f}] "
              f"decided_acc={dec_acc:.4f} cov={cov:.3f}")
bott = pd.DataFrame(rows)
save_tbl(bott, "table09_bottleneck")
pd.DataFrame(preds_long).to_parquet(PROCESSED_DIR / "judgment_evidence_pred.parquet", index=False)
print("[cost] USD:", round(COST.total_usd(), 4))


# 5. Figure: accuracy by evidence condition (the bottleneck curve)
if len(bott):
    fig, ax = plt.subplots(figsize=(6.5, 3.8))
    tiers = list(bott["tier"].unique())
    greys = sns.color_palette("Greys", n_colors=max(len(tiers) + 1, 3))[1:]
    for ti, (tier, g) in enumerate(bott.groupby("tier")):
        g = g.set_index("condition").reindex(CONDS).reset_index()
        y = g["accuracy"].to_numpy()
        lo = y - g["acc_lo"].to_numpy(); hi = g["acc_hi"].to_numpy() - y
        c = greys[ti % len(greys)]
        sns.lineplot(x=g["condition"], y=y, marker="o", color=c, ax=ax, label=tier)
        ax.errorbar(g["condition"], y, yerr=[lo, hi], fmt="none", ecolor=c, capsize=4)
    ax.set_ylabel("judgment accuracy vs gold"); ax.set_ylim(0, 1); ax.set_xlabel("")
    ax.legend(title=""); plt.tight_layout(); save_fig(fig, "fig09_bottleneck"); plt.show()
```
